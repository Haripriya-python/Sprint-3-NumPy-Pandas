# Notebook 8 – Data Manipulation

This notebook covers different techniques for modifying and managing data in Pandas DataFrames.

## Step 1: Understand the Concept

### What Is Renaming Columns?

Renaming columns means changing existing column names in a Pandas DataFrame without changing the actual data.

### Why Do We Use It?

- To make column names clear and meaningful.
- To maintain consistent naming conventions.
- To make data analysis and preprocessing easier.

### Method Used

The `rename()` method changes column names using a dictionary that maps existing names to new names.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a shop maintains a register with headings such as `Item` and `Amount`. Changing `Item` to `Product_Name` makes the information clearer without changing the recorded products.

### Simple Example

| Existing Column | New Column |
|---|---|
| Name | Customer_Name |
| Sales | Total_Sales |

The column headings change, but the values remain the same.

### Business Example

A company can rename `Sales` to `Total_Sales` to make its sales reports easier to understand.

### AI/ML Use Case

During data preprocessing, meaningful column names help data scientists identify input features and prepare datasets for machine learning models.
    

````python
import pandas as pd

# Create a sample customer DataFrame
customer_data = {
    "Name": ["Arun", "Priya", "Ravi"],
    "Sales": [1500, 2200, 1800]
}

customer_df = pd.DataFrame(customer_data)

# Rename the columns
renamed_df = customer_df.rename(
    columns={
        "Name": "Customer_Name",
        "Sales": "Total_Sales"
    }
)

# Display the renamed DataFrame
print(renamed_df)
````

## Step 3: Implement the Concept — Code Explanation

### Code Explanation

1. `import pandas as pd` imports the Pandas library.
2. `pd.DataFrame(customer_data)` creates a DataFrame from the dictionary.
3. `rename(columns={...})` changes the specified column names.
4. The renamed DataFrame is stored in `renamed_df`.
5. `print(renamed_df)` displays the updated column names and the original values.

### What the Code Does

The code changes `Name` to `Customer_Name` and `Sales` to `Total_Sales`. The customer names and sales values remain unchanged.

### Why the Method Is Used

The `rename()` method is useful when existing column names are unclear or need to follow a consistent naming convention. By default, it returns a new DataFrame instead of modifying the original DataFrame.

### Expected Output

```text
  Customer_Name  Total_Sales
0          Arun         1500
1         Priya         2200
2          Ravi         1800
```

The output shows the new column names with the same customer records and sales values.

### Where It Is Used in AI/ML or Data Analytics

Renaming columns helps organize datasets before analysis, visualization, and machine learning preprocessing. Clear names make features easier to identify and work with.

# Drop Rows

## Step 1: Understand the Concept

### What Is Dropping Rows?

Dropping rows means removing selected records from a Pandas DataFrame.

### Why Do We Use It?

- To remove unwanted records from a dataset.
- To exclude records that are not required for a particular analysis.
- To prepare relevant data for further processing.

### Method Used

The `drop()` method removes rows by their index labels. The `index` parameter specifies which row labels should be removed.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a shop register containing five product records. If one record was entered by mistake, the shop manager can remove that record while keeping the remaining records.

### Simple Example

Consider these product records:

| Index | Product | Sales |
|---:|---|---:|
| 0 | Laptop | 50000 |
| 1 | Phone | 20000 |
| 2 | Tablet | 30000 |

If we drop index `1`, the Phone record is removed.

### Business Example

A business may remove an unwanted test transaction from a working analysis dataset after verifying that it should not be included.

### AI/ML Use Case

During data preparation, selected records may be removed when they are confirmed to be invalid or irrelevant. Records should not be removed without checking whether they contain useful information.

````python
import pandas as pd

# Create a sample product DataFrame
product_data = {
    "Product": ["Laptop", "Phone", "Tablet", "Headphones"],
    "Sales": [50000, 20000, 30000, 5000]
}

product_df = pd.DataFrame(product_data)

# Remove the row with index 1
updated_df = product_df.drop(index=1)

# Display the updated DataFrame
print(updated_df)
````

## Step 3: Implement the Concept — Code Explanation

### Code Explanation

1. `import pandas as pd` imports the Pandas library.
2. `pd.DataFrame(product_data)` creates a DataFrame containing product names and sales.
3. `drop(index=1)` removes the row whose index label is `1`.
4. The result is stored in `updated_df`.
5. `print(updated_df)` displays the remaining rows.

### What the Code Does

The code removes the Phone record at index `1`. The Laptop, Tablet, and Headphones records remain in the DataFrame.

### Why the Method Is Used

The `drop()` method is useful for removing specific rows or columns. By default, it returns a new DataFrame and leaves the original DataFrame unchanged.

### Expected Output

```text
       Product  Sales
0        Laptop  50000
2        Tablet  30000
3   Headphones   5000
```

The index labels remain `0`, `2`, and `3` because dropping a row does not automatically reset the index.

### Where It Is Used in AI/ML or Data Analytics

Dropping rows can help prepare datasets by excluding verified invalid records before statistical analysis, visualization, or machine learning. The decision should be based on data quality and the analysis requirements.

# Drop Columns

## Step 1: Understand the Concept

### What Is Dropping Columns?

Dropping columns means removing selected columns from a Pandas DataFrame while keeping the remaining columns.

### Why Do We Use It?

- To remove unnecessary columns.
- To simplify a dataset for analysis.
- To keep only the columns relevant to a particular task.

### Method Used

The `drop()` method removes columns when their names are provided through the `columns` parameter.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a customer form that contains a name, age, purchase amount, and temporary reference number. If the reference number is not needed for a report, we can remove that column while keeping the other information.

### Simple Example

| Name | Age | Purchase | Temporary_ID |
|---|---:|---:|---|
| Arun | 25 | 1500 | T101 |
| Priya | 28 | 2200 | T102 |
| Ravi | 24 | 1800 | T103 |

For purchase analysis, we can remove `Temporary_ID` if it is not required.

### Business Example

A sales team may remove an internal note column before preparing a report containing only product names, quantities, and sales values.

### AI/ML Use Case

During feature preparation, unnecessary columns may be removed when they do not contribute to the task. Relevant information should be retained, and feature selection should be based on the problem and data.

````python
import pandas as pd

# Create a sample customer DataFrame
customer_data = {
    "Name": ["Arun", "Priya", "Ravi"],
    "Age": [25, 28, 24],
    "Purchase": [1500, 2200, 1800],
    "Temporary_ID": ["T101", "T102", "T103"]
}

customer_df = pd.DataFrame(customer_data)

# Remove the Temporary_ID column
updated_df = customer_df.drop(columns=["Temporary_ID"])

# Display the updated DataFrame
print(updated_df)
````

## Step 3: Implement the Concept — Code Explanation

### Code Explanation

1. `import pandas as pd` imports the Pandas library.
2. `pd.DataFrame(customer_data)` creates a DataFrame from the customer data.
3. `drop(columns=["Temporary_ID"])` removes the specified column.
4. The updated DataFrame is stored in `updated_df`.
5. `print(updated_df)` displays the remaining columns.

### What the Code Does

The code removes the `Temporary_ID` column. The `Name`, `Age`, and `Purchase` columns and their values remain unchanged.

### Why the Method Is Used

The `drop()` method helps remove columns that are not required for a specific analysis. By default, it returns a new DataFrame, leaving the original DataFrame unchanged.

### Expected Output

```text id="f3w1ns"
    Name  Age  Purchase
0   Arun   25      1500
1  Priya   28      2200
2   Ravi   24      1800
```

The output contains three columns because `Temporary_ID` has been removed.

### Where It Is Used in AI/ML or Data Analytics

Dropping columns helps prepare a relevant set of features for analysis and model training. Unnecessary identifiers or irrelevant fields may be removed when they are not useful for the task.

# Insert Columns

## Step 1: Understand the Concept

### What Is Inserting Columns?

Inserting columns means adding a new column to a Pandas DataFrame at a specified position.

### Why Do We Use It?

- To add new information to an existing dataset.
- To organize columns in a meaningful order.
- To include additional details without rebuilding the entire DataFrame.

### Method Used

The `insert()` method adds a column at a specified position.

### Syntax

`df.insert(loc, column, value)`

- `loc` specifies the column position.
- `column` specifies the new column name.
- `value` specifies the data to insert.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a school register with Student Name and Age columns. If the school wants to record each student's City between these two columns, it can insert a new City column at that position.

### Simple Example

Before inserting the column:

| Name | Age |
|---|---:|
| Arun | 25 |
| Priya | 28 |

After inserting the City column:

| Name | City | Age |
|---|---|---:|
| Arun | Chennai | 25 |
| Priya | Madurai | 28 |

### Business Example

A company can insert a `Discount` column into its sales DataFrame to store discount amounts for each product.

### AI/ML Use Case

During data preparation, new columns can be added to store useful features, such as a calculated total price or a customer category, before further analysis or model training.

````python
import pandas as pd

# Create a sample customer DataFrame
customer_data = {
    "Name": ["Arun", "Priya", "Ravi"],
    "Age": [25, 28, 24]
}

customer_df = pd.DataFrame(customer_data)

# Insert a City column at position 1
customer_df.insert(
    1,
    "City",
    ["Chennai", "Madurai", "Coimbatore"]
)

# Display the updated DataFrame
print(customer_df)
````

## Step 3: Implement the Concept — Code Explanation

### Code Explanation

1. `import pandas as pd` imports the Pandas library.
2. `pd.DataFrame(customer_data)` creates a DataFrame with Name and Age columns.
3. `customer_df.insert()` adds a new column at the specified position.
4. `1` specifies the second column position.
5. `"City"` is the name of the new column.
6. The list provides one city value for each row.
7. `print(customer_df)` displays the updated DataFrame.

### What the Code Does

The code inserts the City column between Name and Age. Each customer receives the corresponding city value.

### Why the Method Is Used

The `insert()` method is useful when a new column must appear at a specific position. Unlike assigning a column with `df["City"] = values`, `insert()` lets us control where the column appears.

### Expected Output

```text
    Name        City  Age
0   Arun     Chennai   25
1  Priya     Madurai   28
2   Ravi  Coimbatore   24
```

The output shows City as the second column, while the original Name and Age values remain unchanged.

### Where It Is Used in AI/ML or Data Analytics

The `insert()` method can add new features or calculated values to datasets. These features may then be used for reporting, exploratory data analysis, or machine learning preprocessing.

# assign()

## Step 1: Understand the Concept

### What Is `assign()`?

The `assign()` method is used to add new columns to a Pandas DataFrame and return a new DataFrame.

### Why Do We Use It?

- To create new columns easily.
- To create calculated columns.
- To keep the original DataFrame unchanged unless the result is assigned back.
- To create multiple columns in a readable way.

### Method Used

The `assign()` method creates new columns using the column name and its values or expressions.

### Syntax

`df.assign(New_Column=value)`

The new column can also be created using existing columns in the DataFrame.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a shop has a bill containing the quantity and price of each product. The shop wants to calculate the total amount for each product without changing the original bill. A new Total_Sales column can be created for this calculation.

### Simple Example

| Product | Quantity | Price |
|---|---:|---:|
| Laptop | 2 | 50000 |
| Phone | 3 | 20000 |
| Tablet | 1 | 30000 |

A new `Total_Sales` column can be calculated as:

`Quantity × Price`

| Product | Quantity | Price | Total_Sales |
|---|---:|---:|---:|
| Laptop | 2 | 50000 | 100000 |
| Phone | 3 | 20000 | 60000 |
| Tablet | 1 | 30000 | 30000 |

### Business Example

A sales company can use `assign()` to calculate total sales, profit, discount amount, or other derived values from existing columns.

### AI/ML Use Case

In data preprocessing, `assign()` can be used to create new features from existing data. For example, price and quantity can be combined to create a total purchase feature.

````python
import pandas as pd

# Create a sample sales DataFrame
sales_data = {
    "Product": ["Laptop", "Phone", "Tablet"],
    "Quantity": [2, 3, 1],
    "Price": [50000, 20000, 30000]
}

sales_df = pd.DataFrame(sales_data)

# Create a new Total_Sales column
updated_df = sales_df.assign(
    Total_Sales=sales_df["Quantity"] * sales_df["Price"]
)

# Display the updated DataFrame
print(updated_df)
````

## Step 3: Implement the Concept — Code Explanation

### Code Explanation

1. `import pandas as pd` imports the Pandas library.
2. `pd.DataFrame(sales_data)` creates a DataFrame containing Product, Quantity, and Price.
3. `assign()` creates a new column called `Total_Sales`.
4. `sales_df["Quantity"] * sales_df["Price"]` calculates the total sales for each row.
5. The result is stored in `updated_df`.
6. `print(updated_df)` displays the updated DataFrame.

### What the Code Does

The code calculates the total sales for each product by multiplying Quantity by Price and stores the result in a new `Total_Sales` column.

### Why the Method Is Used

The `assign()` method is useful for creating new columns, especially when the new values are calculated from existing columns. It returns a new DataFrame, which helps keep the original DataFrame unchanged.

### Expected Output

```text
  Product  Quantity  Price  Total_Sales
0  Laptop         2  50000       100000
1   Phone         3  20000        60000
2  Tablet         1  30000        30000
```

The `Total_Sales` values are calculated by multiplying Quantity and Price for each product.

### Where It Is Used in AI/ML or Data Analytics

`assign()` is useful for feature engineering and data preparation. New calculated features can be created from existing columns before analysis or machine learning model training.

# apply()

## Step 1: Understand the Concept

### What Is `apply()`?

The `apply()` method is used to apply a function or custom operation to values in a Pandas Series or DataFrame.

### Why Do We Use It?

- To perform custom operations on data.
- To apply the same rule to multiple values.
- To create new values based on conditions or calculations.
- To use user-defined functions with Pandas data.

### Method Used

The `apply()` method applies a function to each value, row, or column depending on how it is used.

### Basic Syntax

`df["Column"].apply(function)`

A lambda function can also be used for simple custom logic.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a school teacher checking students' marks one by one. If a student's mark is 50 or above, the teacher labels the result as "Pass"; otherwise, the result is "Fail". The same rule is applied to every student.

### Simple Example

| Student | Mark |
|---|---:|
| Arun | 80 |
| Priya | 45 |
| Ravi | 70 |

Using a rule:

- Mark >= 50 → Pass
- Mark < 50 → Fail

The result becomes:

| Student | Mark | Result |
|---|---:|---|
| Arun | 80 | Pass |
| Priya | 45 | Fail |
| Ravi | 70 | Pass |

### Business Example

A company can use `apply()` to classify customers based on their purchase amount, such as "High Value", "Medium Value", or "Low Value".

### AI/ML Use Case

`apply()` can be used during feature engineering to transform raw values into meaningful categories or derived features before training a machine learning model.

````python
import pandas as pd

# Create a sample student DataFrame
student_data = {
    "Student": ["Arun", "Priya", "Ravi"],
    "Marks": [80, 45, 70]
}

student_df = pd.DataFrame(student_data)

# Create a Result column using apply()
student_df["Result"] = student_df["Marks"].apply(
    lambda mark: "Pass" if mark >= 50 else "Fail"
)

# Display the updated DataFrame
print(student_df)
````

## Step 3: Implement the Concept — Code Explanation

### Code Explanation

1. `import pandas as pd` imports the Pandas library.
2. `pd.DataFrame(student_data)` creates a DataFrame containing student names and marks.
3. `student_df["Marks"]` selects the Marks column.
4. `apply()` applies the given function to each mark.
5. The lambda function checks whether each mark is greater than or equal to 50.
6. If the condition is true, it returns `"Pass"`.
7. Otherwise, it returns `"Fail"`.
8. The results are stored in a new `Result` column.

### What the Code Does

The code checks every student's mark and creates a result based on the condition.

- 80 → Pass
- 45 → Fail
- 70 → Pass

### Why the Method Is Used

The `apply()` method is useful when we need to apply custom logic that cannot be easily represented using a simple built-in operation.

### Expected Output

```text
  Student  Marks Result
0    Arun     80   Pass
1   Priya     45   Fail
2    Ravi     70   Pass
```

Each student's result is created based on the mark condition.

### Where It Is Used in AI/ML or Data Analytics

`apply()` is commonly used for data transformation and feature engineering. It can convert raw values into categories or create new features based on business or analytical rules.

# map()

## Step 1: Understand the Concept

### What Is `map()`?

The `map()` method is used to transform or replace each value in a Pandas Series based on a mapping, function, or another Series.

### Why Do We Use It?

- To replace values with corresponding values.
- To convert codes into meaningful labels.
- To transform values in a Series.
- To create new columns based on existing values.

### Method Used

The `map()` method applies a mapping to each value in a Pandas Series.

### Basic Syntax

`df["Column"].map(mapping)`

A dictionary can be used when each existing value has a corresponding new value.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a delivery company uses short city codes to identify delivery locations. `CHN` represents Chennai, `MDU` represents Madurai, and `CBE` represents Coimbatore. A mapping system converts these codes into full city names.

### Simple Example

| City Code | City Name |
|---|---|
| CHN | Chennai |
| MDU | Madurai |
| CBE | Coimbatore |

If a DataFrame contains `CHN`, `MDU`, and `CBE`, `map()` can convert them into their corresponding city names.

### Business Example

A company may store product codes such as `P01`, `P02`, and `P03`. A mapping can convert these codes into product names for reports and analysis.

### AI/ML Use Case

`map()` can be used during preprocessing to convert categorical labels or codes into meaningful values before analysis or feature engineering.

````python
import pandas as pd

# Create a sample customer DataFrame
customer_data = {
    "Name": ["Arun", "Priya", "Ravi"],
    "City_Code": ["CHN", "MDU", "CBE"]
}

customer_df = pd.DataFrame(customer_data)

# Create a mapping dictionary
city_mapping = {
    "CHN": "Chennai",
    "MDU": "Madurai",
    "CBE": "Coimbatore"
}

# Map city codes to city names
customer_df["City"] = customer_df["City_Code"].map(city_mapping)

# Display the updated DataFrame
print(customer_df)
````

## Step 3: Implement the Concept — Code Explanation

### Code Explanation

1. `import pandas as pd` imports the Pandas library.
2. `pd.DataFrame(customer_data)` creates a customer DataFrame.
3. `city_mapping` contains the relationship between city codes and city names.
4. `customer_df["City_Code"]` selects the City_Code column.
5. `map(city_mapping)` finds the corresponding city name for each code.
6. The mapped values are stored in a new `City` column.

### What the Code Does

The code converts each city code into its corresponding city name.

- `CHN` → Chennai
- `MDU` → Madurai
- `CBE` → Coimbatore

### Why the Method Is Used

The `map()` method is useful when each value needs to be replaced or transformed according to a known mapping. A dictionary makes these value-to-value relationships easy to define.

### Expected Output

```text
    Name City_Code        City
0   Arun       CHN     Chennai
1  Priya       MDU     Madurai
2   Ravi       CBE  Coimbatore
```

The new `City` column contains the city names obtained from the mapping dictionary.

### Where It Is Used in AI/ML or Data Analytics

`map()` can be used for categorical data transformation, label conversion, code-to-name mapping, and feature preparation before analysis or machine learning.

# replace()

## Step 1: Understand the Concept

### What Is `replace()`?

The `replace()` method is used to replace specific values in a Pandas Series or DataFrame with new values.

### Why Do We Use It?

- To correct incorrect values.
- To standardize inconsistent values.
- To replace short codes with meaningful values.
- To update specific values in a dataset.

### Method Used

The `replace()` method searches for specified values and replaces them with the given new values.

### Basic Syntax

`df["Column"].replace(old_value, new_value)`

A dictionary can also be used to replace multiple values at the same time.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a customer form where gender is recorded using `M` and `F`. For a report, the company wants to display `Male` and `Female` instead. The short values can be replaced with meaningful values.

### Simple Example

| Existing Value | New Value |
|---|---|
| M | Male |
| F | Female |

For example, if the dataset contains `M`, `F`, and `M`, the values can be replaced with `Male`, `Female`, and `Male`.

### Business Example

A company may have product status values such as `A` and `I`. These can be replaced with `Active` and `Inactive` to make business reports easier to understand.

### AI/ML Use Case

`replace()` can be used during data cleaning to standardize categorical values before analysis or machine learning preprocessing.

````python
import pandas as pd

# Create a sample customer DataFrame
customer_data = {
    "Name": ["Arun", "Priya", "Ravi", "Divya"],
    "Gender": ["M", "F", "M", "F"]
}

customer_df = pd.DataFrame(customer_data)

# Replace short gender codes with meaningful values
customer_df["Gender"] = customer_df["Gender"].replace({
    "M": "Male",
    "F": "Female"
})

# Display the updated DataFrame
print(customer_df)
````

## Step 3: Implement the Concept — Code Explanation

### Code Explanation

1. `import pandas as pd` imports the Pandas library.
2. `pd.DataFrame(customer_data)` creates a customer DataFrame.
3. The `Gender` column contains the values `M` and `F`.
4. `replace()` uses a dictionary to define the old and new values.
5. `M` is replaced with `Male`.
6. `F` is replaced with `Female`.
7. The updated values are stored back in the `Gender` column.

### What the Code Does

The code replaces short gender codes with their full names.

- `M` → Male
- `F` → Female

### Why the Method Is Used

The `replace()` method is useful for cleaning and standardizing values in a dataset. It is especially useful when known values need to be changed into consistent or meaningful values.

### Expected Output

```text
    Name  Gender
0   Arun    Male
1  Priya  Female
2   Ravi    Male
3  Divya  Female
```

The `Gender` column now contains the full values instead of the short codes.

### Where It Is Used in AI/ML or Data Analytics

`replace()` is commonly used during data cleaning and preprocessing to standardize categorical values and correct known data inconsistencies before analysis or model training.

# Duplicate Handling

## Step 1: Understand the Concept

### What Is Duplicate Handling?

Duplicate handling means identifying and removing repeated records from a dataset.

A duplicate row contains the same values as another row for the columns being considered.

### Why Do We Handle Duplicates?

- To avoid repeated records.
- To improve data quality.
- To prevent incorrect counts and calculations.
- To prepare cleaner data for analysis and machine learning.

### Methods Used

`duplicated()` identifies duplicate rows and returns Boolean values.

`drop_duplicates()` removes duplicate rows from a DataFrame.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a customer register where the same customer entry was accidentally written twice. Before preparing the final report, the repeated entry should be identified and removed.

### Simple Example

| Name | City | Purchase |
|---|---|---:|
| Arun | Chennai | 1500 |
| Priya | Madurai | 2200 |
| Arun | Chennai | 1500 |
| Ravi | Coimbatore | 1800 |

The Arun record appears twice, so it is a duplicate.

### Business Example

An e-commerce company may receive the same order record multiple times due to a data entry or system issue. Duplicate handling prevents the same order from being counted twice in sales reports.

### AI/ML Use Case

Duplicate records can give repeated importance to the same observation during model training. Removing unnecessary duplicates can help create a cleaner and more reliable training dataset.

````python
import pandas as pd

# Create a DataFrame with duplicate records
customer_data = {
    "Name": ["Arun", "Priya", "Arun", "Ravi"],
    "City": ["Chennai", "Madurai", "Chennai", "Coimbatore"],
    "Purchase": [1500, 2200, 1500, 1800]
}

customer_df = pd.DataFrame(customer_data)

# Identify duplicate rows
print("Duplicate rows:")
print(customer_df.duplicated())

# Remove duplicate rows
cleaned_df = customer_df.drop_duplicates()

# Display the cleaned DataFrame
print("\nDataFrame after removing duplicates:")
print(cleaned_df)
````

## Step 3: Implement the Concept — Code Explanation

### Code Explanation

1. `import pandas as pd` imports the Pandas library.
2. `pd.DataFrame(customer_data)` creates a DataFrame containing customer records.
3. `duplicated()` checks whether each row is a duplicate of a previous row.
4. It returns `True` for a duplicate row and `False` for a non-duplicate row.
5. `drop_duplicates()` removes repeated rows.
6. The cleaned result is stored in `cleaned_df`.
7. `print()` displays the duplicate information and cleaned DataFrame.

### What the Code Does

The code first identifies duplicate rows and then removes the repeated Arun record.

### Why the Methods Are Used

`duplicated()` is useful for identifying repeated records before deciding how to handle them.

`drop_duplicates()` is useful when duplicate records need to be removed from the DataFrame.

### Expected Output

```text
Duplicate rows:
0    False
1    False
2     True
3    False
dtype: bool

DataFrame after removing duplicates:
    Name        City  Purchase
0   Arun     Chennai      1500
1  Priya     Madurai      2200
3   Ravi  Coimbatore      1800
```

The third row is marked `True` because it is a duplicate of the first row. After using `drop_duplicates()`, only one copy of the Arun record remains.

### Where It Is Used in AI/ML or Data Analytics

Duplicate handling is an important data-cleaning step. It helps prevent repeated observations from affecting statistical analysis, reports, and machine learning datasets.
