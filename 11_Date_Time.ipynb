# Datetime

## Step 1: Understand the Concept

Datetime represents both date and time in a single value.

A datetime value can contain:
- Year
- Month
- Day
- Hour
- Minute
- Second

Example:
2026-10-08 21:00:00

Here:
- 2026 represents the year.
- 10 represents the month.
- 08 represents the day.
- 21:00:00 represents the time.

### Why is Datetime Used?

Datetime is used when we need to store, analyze, compare, or calculate dates and times.

Common examples include:
- Customer order date and time
- Bank transaction time
- Employee joining date
- Website login time
- Sensor readings

### Common Pandas Function

pd.to_datetime() is used to convert date or time values into Pandas datetime format.

Syntax:

pd.to_datetime(data)



## Step 2: Demonstrate the Concept

### Real-World Analogy

Think of a train ticket.

A ticket may show:
- Travel Date: 08 October 2026
- Departure Time: 09:30 AM

When date and time are considered together, they form a datetime value.

### Simple Example

Suppose a customer places an order at:

2026-10-08 10:30:00

This tells us exactly when the order happened.

| Date Part | Value |
|---|---|
| Year | 2026 |
| Month | 10 |
| Day | 08 |
| Time | 10:30:00 |

### Business Example

An e-commerce company stores the date and time of every customer order.

Using datetime values, the company can analyze:
- When orders are placed
- Which hours have more orders
- Which days have higher sales
- Daily or monthly sales trends

### AI/ML Use Case

In a sales prediction model, order dates can be converted into datetime values.

From the datetime value, we can later extract useful features such as:
- Year
- Month
- Day
- Day of the week
- Hour

These features can help an ML model identify time-based patterns.

```python
import pandas as pd

# Create a date and time value
order_date = pd.to_datetime("2026-10-08 10:30:00")

print(order_date)
print(type(order_date))
```

## Step 3: Code Explanation

### Code Explanation

First, we import Pandas using the alias `pd`.

Then, `pd.to_datetime()` converts the given date and time string into a Pandas datetime value.

### What the Code Does

The code converts:

"2026-10-08 10:30:00"

into a Pandas datetime object.

The `print()` statements display:
- The converted datetime value
- Its data type

### Why the Function Is Used

`pd.to_datetime()` is used to convert strings or other date-like values into a proper datetime format.

This makes it easier to perform date and time operations later.

The output will be similar to:

The first line shows the converted date and time.

# Timestamp

## Step 1: Understand the Concept

A Timestamp represents a specific date and time.

In Pandas, a Timestamp is used to represent a single point in time.

Example:

2026-10-08 10:30:00

A Timestamp can contain:
- Year
- Month
- Day
- Hour
- Minute
- Second

Pandas provides `pd.Timestamp()` to create a Timestamp directly.

Syntax:

pd.Timestamp("YYYY-MM-DD HH:MM:SS")

A Timestamp can also be created using `pd.to_datetime()`.

    

## Step 2: Demonstrate the Concept

### Real-World Analogy

Think about a receipt from a shop.

The receipt may show:

08 October 2026, 10:30 AM

This tells us the exact point in time when the purchase happened.

That exact date and time can be represented as a Timestamp.

### Simple Example

Suppose a customer places an online order at:

2026-10-08 10:30:00

The Timestamp tells us exactly when the order was placed.

### Business Example

An e-commerce company records the exact time when each order is placed.

Using Timestamps, the company can:
- Compare order times
- Find the earliest order
- Find the latest order
- Calculate the time between orders

### AI/ML Use Case

In time-based datasets, Timestamps help organize events in chronological order.

For example, a sales dataset may contain:

| Timestamp | Sales |
|---|---:|
| 2026-10-08 09:00:00 | 500 |
| 2026-10-08 10:00:00 | 750 |
| 2026-10-08 11:00:00 | 900 |

This type of data can be used for time-series analysis and forecasting.

```python
import pandas as pd

# Create a Pandas Timestamp
timestamp = pd.Timestamp("2026-10-08 10:30:00")

print(timestamp)
print(type(timestamp))

# Access individual date and time components
print("Year:", timestamp.year)
print("Month:", timestamp.month)
print("Day:", timestamp.day)
print("Hour:", timestamp.hour)
```

## Step 3: Code Explanation

### Code Explanation

`pd.Timestamp()` creates a Pandas Timestamp from the given date and time.

The `type()` function shows the data type of the created value.

The attributes `.year`, `.month`, `.day`, and `.hour` are used to access individual parts of the Timestamp.

### What the Code Does

The code:
1. Creates a Timestamp for 8 October 2026 at 10:30 AM.
2. Displays the Timestamp.
3. Displays its data type.
4. Extracts the year, month, day, and hour.

### Why the Function and Attributes Are Used

`pd.Timestamp()` is used to create a specific point in time.

`.year` returns the year.

`.month` returns the month.

`.day` returns the day.

`.hour` returns the hour.

These attributes are useful when we need specific date or time information for analysis.

### Expected Output

The output will be:

2026-10-08 10:30:00
<class 'pandas._libs.tslibs.timestamps.Timestamp'>
Year: 2026
Month: 10
Day: 8
Hour: 10

The output shows the complete Timestamp and the individual date and time components.

### Where It Is Used in AI/ML or Data Analytics

- Time-series analysis

- Time-based feature engineering

# Timedelta

## Step 1: Understand the Concept

Timedelta represents a duration or the difference between two dates or times.

For example:

Order Date: 2026-10-08
Delivery Date: 2026-10-11
Duration: 3 days

Pandas supports time difference calculations by subtracting datetime values.

The `pd.Timedelta()` function can also create a specific duration.

Syntax:

pd.Timedelta(days=3)

Common Timedelta units include:
- days
- hours
- minutes
- seconds

Timedelta is useful for measuring how much time has passed between two events.



## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine ordering a product online.

You place the order on Monday, and it arrives on Thursday.

The difference between the order date and delivery date is 3 days.

This duration is a Timedelta.

### Simple Example

Suppose a product is ordered on 8 October 2026 and delivered on 11 October 2026.

| Event | Date |
|---|---|
| Order Placed | 2026-10-08 |
| Product Delivered | 2026-10-11 |
| Delivery Duration | 3 days |

### Business Example

An e-commerce company calculates delivery duration for customer orders.

This helps the company:
- Measure delivery performance
- Identify delayed orders
- Compare delivery times
- Improve logistics planning

### AI/ML Use Case

In delivery prediction, the difference between order time and delivery time can help create duration-based features.

Historical delivery durations can be analyzed to understand patterns and may help a model estimate future delivery times.

```python
import pandas as pd

# Create order and delivery dates
order_date = pd.Timestamp("2026-10-08")
delivery_date = pd.Timestamp("2026-10-11")

# Calculate the difference between the dates
delivery_duration = delivery_date - order_date

print("Order Date:", order_date.date())
print("Delivery Date:", delivery_date.date())
print("Delivery Duration:", delivery_duration)
print("Number of Days:", delivery_duration.days)

# Create a duration of 5 days
extra_days = pd.Timedelta(days=5)
print("Extra Duration:", extra_days)
```

## Step 3: Code Explanation

### Code Explanation

`pd.Timestamp()` creates date values.

Subtracting `order_date` from `delivery_date` calculates the time difference and returns a Timedelta.

The `.days` attribute extracts the number of whole days from the duration.

`pd.Timedelta(days=5)` creates a duration of five days.

### What the Code Does

The code:
1. Creates the order date and delivery date.
2. Subtracts the order date from the delivery date.
3. Displays the duration.
4. Extracts the number of days.
5. Creates and displays another duration of five days.

### Why the Functions and Attributes Are Used

- `pd.Timestamp()` creates a specific date and time.
- Subtraction calculates the difference between two dates.
- `.days` returns the number of whole days in the duration.
- `pd.Timedelta()` creates a duration that can be added to or subtracted from a date.

### Expected Output

Order Date: 2026-10-08
Delivery Date: 2026-10-11
Delivery Duration: 3 days 00:00:00
Number of Days: 3
Extra Duration: 5 days 00:00:00

The output shows that delivery took three days and that the additional duration represents five days.

### Where It Is Used in AI/ML or Data Analytics

Timedelta is used in:
- Delivery time analysis
- Customer response time analysis
- Employee attendance analysis
- Transaction duration analysis
- Time-series data preprocessing
- Delivery time prediction

# Time Zones

## Step 1: Understand the Concept

A time zone represents the local time used in a particular region of the world.

Different locations may have different local times at the same moment.

India uses Indian Standard Time (IST), which is UTC+05:30.

Pandas supports timezone-aware datetime values.

Common operations include:

- `pd.Timestamp()` to create a timestamp.
- `tz_localize()` to assign a time zone to a timezone-naive timestamp.
- `tz_convert()` to convert a timezone-aware timestamp to another time zone.

Example time zones:

- Asia/Kolkata
- America/New_York
- Europe/London

A timezone-naive timestamp does not contain time zone information. A timezone-aware timestamp includes time zone information.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine scheduling an online meeting with a colleague in another country.

You plan the meeting for 10:00 AM in India.

Your colleague needs to know the corresponding local time in their country.

Time zone conversion helps represent the same moment in both locations.

### Simple Example

A timestamp can be created in the India time zone and converted to another time zone.

| Location | Time Zone |
|---|---|
| Chennai | Asia/Kolkata |
| New York | America/New_York |
| London | Europe/London |

The local clock time may differ, but the converted timestamps represent the same moment.

### Business Example

A company has customers in India, the United States, and the United Kingdom.

Each customer may place an order using local time.

Time zone handling helps the company compare order times consistently and generate accurate reports.

### AI/ML Use Case

Time zone handling is useful when preparing global time-series datasets.

For example, an application can convert events to a common time zone before analyzing customer activity across different countries.

This helps avoid incorrect comparisons caused by mixing local times.

```python
import pandas as pd

# Create a timestamp in the India time zone
india_time = pd.Timestamp(
    "2026-10-08 10:00:00",
    tz="Asia/Kolkata"
)

# Convert it to the New York time zone
new_york_time = india_time.tz_convert("America/New_York")

print("India Time:", india_time)
print("New York Time:", new_york_time)

# Display the time zone information
print("India Time Zone:", india_time.tz)
print("New York Time Zone:", new_york_time.tz)
```

## Step 3: Code Explanation

### Code Explanation

`pd.Timestamp()` creates a timestamp.

The `tz` parameter assigns the Asia/Kolkata time zone when creating the timestamp.

`tz_convert()` converts the timezone-aware timestamp to another time zone while preserving the same moment in time.

### What the Code Does

The code:
1. Creates a timestamp for 10:00 AM in India.
2. Converts the timestamp to New York time.
3. Displays both timestamps.
4. Displays the time zone information for each timestamp.

### Why the Functions and Parameters Are Used

- `pd.Timestamp()` creates a specific date and time.
- `tz="Asia/Kolkata"` specifies the India time zone.
- `tz_convert()` converts a timezone-aware timestamp to another time zone.
- `.tz` returns the time zone information.

### Expected Output

The output will be similar to:

India Time: 2026-10-08 10:00:00+05:30
New York Time: 2026-10-08 00:30:00-04:00
India Time Zone: Asia/Kolkata
New York Time Zone: America/New_York

New York is on daylight saving time on this date, so its local time is 9 hours and 30 minutes behind India.

Both timestamps represent the same moment in time.

### Where It Is Used in AI/ML or Data Analytics

Time zones are used in:
- Global customer analytics
- International transaction analysis
- Website activity analysis
- Time-series preprocessing
- Global sales reporting
- Event data from distributed systems

# Extracting Date Features

## Step 1: Understand the Concept

Extracting date features means obtaining useful components from a datetime value.

Common date features include:

- Year
- Month
- Day
- Hour
- Minute
- Second
- Day of the week
- Quarter

In Pandas, the `.dt` accessor is used to access date and time components from a datetime Series.

Common attributes:

- `.dt.year` extracts the year.
- `.dt.month` extracts the month.
- `.dt.day` extracts the day.
- `.dt.hour` extracts the hour.
- `.dt.dayofweek` extracts the weekday as a number, where Monday is 0 and Sunday is 6.
- `.dt.quarter` extracts the quarter of the year.

Syntax:

dataframe["Date_Column"].dt.year

Date features are useful for understanding patterns and preparing data for analysis and machine learning.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a calendar showing the date 8 October 2026.

From this date, you can identify the year, month, day, and weekday without needing a separate calendar entry for each value.

Extracting date features performs this task using Python.

### Simple Example

Consider the datetime value:

2026-10-08 14:30:00

| Feature | Value |
|---|---|
| Year | 2026 |
| Month | 10 |
| Day | 8 |
| Hour | 14 |
| Weekday | Thursday |
| Quarter | 4 |

### Business Example

An e-commerce company stores the date and time of every customer order.

By extracting the month and weekday, the company can analyze:
- Monthly sales trends
- Weekday order patterns
- Busy shopping hours
- Seasonal demand

### AI/ML Use Case

In sales forecasting, raw dates can be converted into separate features such as month, weekday, and hour.

A machine learning model can use these features to learn time-based patterns in historical sales data.

This process is called feature engineering.

```python
import pandas as pd

# Create a sample sales DataFrame
sales_df = pd.DataFrame({
    "Order_Date": [
        "2026-10-08 14:30:00",
        "2026-10-09 10:15:00",
        "2026-10-10 18:45:00"
    ],
    "Sales": [5000, 7500, 6200]
})

# Convert the Order_Date column into datetime format
sales_df["Order_Date"] = pd.to_datetime(sales_df["Order_Date"])

# Extract date features
sales_df["Year"] = sales_df["Order_Date"].dt.year
sales_df["Month"] = sales_df["Order_Date"].dt.month
sales_df["Day"] = sales_df["Order_Date"].dt.day
sales_df["Hour"] = sales_df["Order_Date"].dt.hour
sales_df["Weekday"] = sales_df["Order_Date"].dt.day_name()
sales_df["Quarter"] = sales_df["Order_Date"].dt.quarter

print(sales_df)
```

## Step 3: Code Explanation

### Code Explanation

First, a DataFrame is created with order dates and sales values.

`pd.to_datetime()` converts the Order_Date column into datetime format.

The `.dt` accessor extracts individual date features from the datetime column.

Each extracted feature is stored in a new DataFrame column.

### What the Code Does

The code:
1. Creates a sample sales dataset.
2. Converts order dates into datetime values.
3. Extracts the year, month, day, hour, weekday name, and quarter.
4. Adds these features as new columns.
5. Displays the updated DataFrame.

### Why the Functions and Attributes Are Used

- `pd.to_datetime()` converts strings into datetime values.
- `.dt.year` extracts the year.
- `.dt.month` extracts the month.
- `.dt.day` extracts the day of the month.
- `.dt.hour` extracts the hour.
- `.dt.day_name()` returns the weekday name.
- `.dt.quarter` returns the quarter of the year.

These features help us analyze date-based patterns without manually examining each date.

### Expected Output

The output will contain the original Order_Date and Sales columns, along with the new Year, Month, Day, Hour, Weekday, and Quarter columns.

For example, the first row will have:

- Year: 2026
- Month: 10
- Day: 8
- Hour: 14
- Weekday: Thursday
- Quarter: 4

The new columns make the sales data easier to analyze by date and time.

### Where It Is Used in AI/ML or Data Analytics

Extracting date features is used in:
- Sales forecasting
- Customer behavior analysis
- Demand prediction
- Time-series analysis
- Seasonal trend analysis
- Machine learning feature engineering
