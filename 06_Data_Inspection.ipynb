## Topic 1: head()

## Step 1: Understand the Concept

The `head()` method displays the first few rows of a Pandas DataFrame.

By default, `df.head()` returns the first five rows. We can specify a number inside the parentheses to display a different number of rows.

It is useful for previewing a dataset and checking whether the data has loaded correctly.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine opening a notebook and looking at the first five entries to understand what information it contains. The `head()` method does the same thing with a DataFrame.

### Simple Example

A teacher has marks for 100 students. Using `head()` displays the first five student records without showing all 100 rows.

### Business Example

A sales analyst loads thousands of sales records and uses `head()` to preview the product names, quantities, and sales values.

### AI/ML Use Case

An AI/ML developer uses `head()` after loading a dataset to inspect sample records and verify that the expected features and values are present.

````python
# Import Pandas
import pandas as pd

# Create sample customer data
customer_data = {
    "Name": ["Arun", "Priya", "Ravi", "Divya", "Kumar", "Meena"],
    "Age": [25, 28, 24, 30, 27, 26],
    "Purchase": [1500, 2200, 1800, 3000, 2500, 1700]
}

# Create a DataFrame
customer_df = pd.DataFrame(customer_data)

# Display the first five rows
print(customer_df.head())

# Display the first three rows
print(customer_df.head(3))
````

## Step 3: Implement the Concept

### Code Explanation

- `import pandas as pd` imports the Pandas library.
- `customer_data` stores sample customer information in a dictionary.
- `pd.DataFrame(customer_data)` converts the dictionary into a DataFrame.
- `customer_df.head()` displays the first five rows.
- `customer_df.head(3)` displays the first three rows.

### What the Code Does

The code creates a customer dataset containing six records and displays the first five rows, followed by the first three rows.

### Why the Function/Method Is Used

The `head()` method helps us inspect the beginning of a dataset quickly without printing every record. The number inside the parentheses controls how many rows are displayed.

### Expected Output

For `customer_df.head()`:

```text
    Name  Age  Purchase
0   Arun   25      1500
1  Priya   28      2200
2   Ravi   24      1800
3  Divya   30      3000
4  Kumar   27      2500
```

For `customer_df.head(3)`, only Arun, Priya, and Ravi are displayed.

The first command shows five rows, while the second shows three rows.

### Where It Is Used in AI/ML or Data Analytics

The `head()` method is commonly used after loading a dataset to preview records, check feature names, and identify obvious data issues before data cleaning or model training.

## Topic 3: sample()

## Step 1: Understand the Concept

The `sample()` method randomly selects rows from a Pandas DataFrame.

By default, `df.sample()` returns one random row. We can specify the number of rows using `df.sample(n)`.

The `frac` parameter can be used to select a fraction of the DataFrame.

For example, `df.sample(3)` selects three random rows, while `df.sample(frac=0.5)` selects 50% of the rows.

The selected rows may be different each time because the method uses random selection.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine picking a few names randomly from a box containing many names. The `sample()` method performs a similar random selection from a DataFrame.

### Simple Example

A teacher has records for 100 students and wants to randomly check five student records. The teacher can use `sample(5)`.

### Business Example

A company has thousands of customer records. A data analyst can randomly select a few records to inspect the data without checking every row.

### AI/ML Use Case

An AI/ML developer can randomly inspect records from a large dataset or select a random subset of data for analysis and experimentation.

````python
# Import Pandas
import pandas as pd

# Create sample customer data
customer_data = {
    "Name": ["Arun", "Priya", "Ravi", "Divya", "Kumar", "Meena"],
    "Age": [25, 28, 24, 30, 27, 26],
    "Purchase": [1500, 2200, 1800, 3000, 2500, 1700]
}

# Create a DataFrame
customer_df = pd.DataFrame(customer_data)

# Select one random row
print("One random row:")
print(customer_df.sample())

# Select three random rows
print("\nThree random rows:")
print(customer_df.sample(3))
````

## Step 3: Implement the Concept

### Code Explanation

- `import pandas as pd` imports the Pandas library.
- `customer_data` stores sample customer information.
- `pd.DataFrame(customer_data)` converts the dictionary into a DataFrame.
- `customer_df.sample()` selects one random row.
- `customer_df.sample(3)` selects three random rows.
- `print()` displays the selected records.

### What the Code Does

The code creates a customer DataFrame and randomly selects one row and three rows from it.

### Why the Function/Method Is Used

The `sample()` method is used when we want to inspect randomly selected records instead of always looking at the beginning or end of the dataset.

The number inside `sample()` specifies how many rows should be selected.

### Expected Output

The output will contain randomly selected rows, so the exact rows may change each time the code is executed.

For example:

```text
One random row:
    Name  Age  Purchase
4  Kumar   27      2500

Three random rows:
    Name  Age  Purchase
1  Priya   28      2200
5  Meena   26      1700
2   Ravi   24      1800
```

The actual rows and their order can be different because the selection is random.

### Where It Is Used in AI/ML or Data Analytics

The `sample()` method is useful for randomly inspecting large datasets, creating random subsets for analysis, and checking whether the data contains a reasonable variety of records.

## Topic 5: describe()

## Step 1: Understand the Concept

The `describe()` method generates a statistical summary of numerical columns in a Pandas DataFrame.

It provides values such as count, mean, standard deviation, minimum, quartiles, and maximum.

The quartiles are represented by 25%, 50%, and 75%. The 50% value represents the median.

The `describe()` method is useful for quickly understanding the distribution and range of numerical data.

## Step 2: Demonstrate the Concept

### Real-World Analogy

### Business Example

### AI/ML Use Case

An AI/ML developer uses `describe()` during exploratory data analysis to understand numerical features, identify unusual ranges, and get an initial understanding of the dataset before preprocessing and model training.

````python
# Import Pandas
import pandas as pd

# Create customer data
customer_data = {
    "Name": ["Arun", "Priya", "Ravi", "Divya", "Kumar"],
    "Age": [20, 25, 30, 35, 40],
    "Purchase": [1000, 2000, 3000, 4000, 5000]
}

# Create a DataFrame
customer_df = pd.DataFrame(customer_data)

# Display statistical summary
print(customer_df.describe())
````

### What the Code Does

### Why the Function/Method Is Used

The `describe()` method is used to quickly understand numerical data without calculating each statistical value separately.

It helps us identify the center, spread, range, and distribution of numerical features.

# Topic 6 - `shape`

## Step 1 – Understand the Concept

The `shape` attribute tells us the number of rows and columns in a DataFrame.

It returns the result in the format:

(rows, columns)

For example, `(5, 3)` means:
- 5 rows
- 3 columns

We can also use:
- `df.shape[0]` to get the number of rows
- `df.shape[1]` to get the number of columns

`shape` is an attribute, so we do not use parentheses.

## Step 2 – Demonstrate the Concept

### Real-World Analogy

Think of a classroom seating arrangement. If there are 5 students and 3 details are recorded for each student, the table has 5 rows and 3 columns.

### Simple Example

A DataFrame with 5 customer records and 3 columns has a shape of `(5, 3)`.

### Business Example

A company may have a customer dataset containing thousands of customer records and several columns such as age, location, purchase amount, and membership type. The `shape` attribute helps quickly check the size of the dataset.

### AI/ML Use Case

In AI/ML, `shape` is used to understand the size of the dataset before training a model. It helps identify the number of records and features available as model input.

````python
# Import Pandas
import pandas as pd

# Create customer data
customer_data = {
    "Name": ["Arun", "Priya", "Ravi", "Divya", "Kumar"],
    "Age": [20, 25, 30, 35, 40],
    "Purchase": [1000, 2000, 3000, 4000, 5000]
}

# Create a DataFrame
customer_df = pd.DataFrame(customer_data)

# Display the shape of the DataFrame
print(customer_df.shape)

# Display the number of rows
print("Number of rows:", customer_df.shape[0])

# Display the number of columns
print("Number of columns:", customer_df.shape[1])
````

## Step 3 – Code Explanation

### Code Explanation

- `import pandas as pd` imports the Pandas library.
- `customer_data` stores customer information.
- `pd.DataFrame(customer_data)` creates the DataFrame.
- `customer_df.shape` returns the number of rows and columns.
- `customer_df.shape[0]` returns the number of rows.
- `customer_df.shape[1]` returns the number of columns.

### What the Code Does

The code creates a DataFrame with 5 customer records and 3 columns. It then displays the complete shape, number of rows, and number of columns.

### Why the Attribute Is Used

The `shape` attribute is used to quickly understand the size of a DataFrame.

It is useful for checking:
- How many records are available
- How many columns are available
- How many features are present in a dataset

### Expected Output

```text
(5, 3)
Number of rows: 5
Number of columns: 3

The result (5, 3) means the DataFrame contains 5 rows and 3 columns.

### Where It Is Used in AI/ML or Data Analytics

In AI/ML, shape helps check dataset size and the number of features before data preprocessing and model training. It can also help verify whether the data has the expected number of rows and columns.

# Topic - nunique()

## Step 1 – Understand the Concept

The `nunique()` method returns the number of unique values in a DataFrame column.

A unique value is a value that is different from the others.

For example, if a column contains:

20, 25, 20, 30, 25

The unique values are:

20, 25, 30

So, `nunique()` returns `3`.

`nunique()` is a method, so we use parentheses.

## Step 2 – Demonstrate the Concept

### Real-World Analogy

Think of a shopping store where the same product may be purchased many times. If Laptop appears 10 times and Phone appears 5 times, there are still only 2 unique products.

### Simple Example

Suppose we have the following customer data:

| Customer | City |
|----------|------|
| Arun | Chennai |
| Priya | Madurai |
| Ravi | Chennai |
| Divya | Coimbatore |
| Kumar | Madurai |

The City column contains repeated values, but the unique cities are Chennai, Madurai, and Coimbatore.

Therefore, `nunique()` for the City column returns `3`.

### Business Example

A company can use `nunique()` to find the number of unique products, cities, customers, or categories in a dataset.

### AI/ML Use Case

In AI/ML, `nunique()` helps identify the number of unique values in categorical features. This can help decide how a categorical column should be handled during preprocessing.

````python
# Import Pandas
import pandas as pd

# Create customer data
customer_data = {
    "Name": ["Arun", "Priya", "Ravi", "Divya", "Kumar"],
    "City": ["Chennai", "Madurai", "Chennai", "Coimbatore", "Madurai"],
    "Purchase": [1500, 2200, 1800, 3000, 2500]
}

# Create a DataFrame
customer_df = pd.DataFrame(customer_data)

# Count unique cities
unique_cities = customer_df["City"].nunique()

# Display the number of unique cities
print("Number of unique cities:", unique_cities)
````

## Step 3 – Code Explanation

### Code Explanation

- `import pandas as pd` imports the Pandas library.
- `customer_data` stores customer information.
- `pd.DataFrame(customer_data)` creates the DataFrame.
- `customer_df["City"]` selects the City column.
- `.nunique()` counts the different values in the City column.
- The result is stored in `unique_cities`.

### What the Code Does

The code creates a customer DataFrame and counts how many different cities are present in the City column.

### Why the Method Is Used

The `nunique()` method is used to find the number of unique values in a column.

It is useful for:
- Finding the number of categories
- Identifying repeated values
- Understanding categorical data
- Data exploration and preprocessing

### Expected Output

```text
Number of unique cities: 3

There are three unique cities: Chennai, Madurai, and Coimbatore.

### Where It Is Used in AI/ML or Data Analytics

In AI/ML and Data Analytics, nunique() is commonly used during exploratory data analysis to understand categorical features and determine how many different categories they contain.

### Business Example

A company can use `value_counts()` to find which products are purchased most often, which cities have the most customers, or which customer categories are most common.

### AI/ML Use Case

In AI/ML, `value_counts()` helps understand categorical features and their frequency. It can also help identify highly imbalanced categories before model training.

````python
# Import Pandas
import pandas as pd

# Create customer data
customer_data = {
    "Name": ["Arun", "Priya", "Ravi", "Divya", "Kumar", "Meena"],
    "City": ["Chennai", "Madurai", "Chennai", "Coimbatore", "Madurai", "Chennai"]
}

# Create a DataFrame
customer_df = pd.DataFrame(customer_data)

# Count the occurrences of each city
city_counts = customer_df["City"].value_counts()

# Display the counts
print(city_counts)
````

## Step 3 – Code Explanation

### Code Explanation

- `import pandas as pd` imports the Pandas library.
- `customer_data` stores customer information.
- `pd.DataFrame(customer_data)` creates the DataFrame.
- `customer_df["City"]` selects the City column.
- `.value_counts()` counts how many times each city appears.
- The result is stored in `city_counts`.

### What the Code Does

The code creates a customer DataFrame and counts the number of customers from each city.

### Why the Method Is Used

The `value_counts()` method is used to understand the frequency of each unique value in a column.

It is useful for:
- Finding the most common values
- Understanding categorical data
- Checking category distribution
- Identifying imbalanced categories
