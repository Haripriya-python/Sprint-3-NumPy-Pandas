# Topic 1 – CSV Files

## Step 1 – Understand the Concept

CSV stands for Comma-Separated Values.

A CSV file stores tabular data in rows and columns. Each row represents a record, and commas are commonly used to separate the values in different columns.

CSV files are simple, lightweight, and commonly used for storing and exchanging datasets.

Commonly used Pandas functions:
- `pd.read_csv()` – reads data from a CSV file.
- `df.to_csv()` – writes a DataFrame to a CSV file.

CSV files are widely used in Data Analytics and AI/ML projects because datasets are often available in CSV format.

## Step 2 – Demonstrate the Concept

### Real-World Analogy

Think of a table in a notebook containing customer names, ages, and purchases. A CSV file stores similar tabular information in a simple text format, with commas separating the columns.

### Simple Example

A CSV file may contain:

Name,Age,Purchase

Arun,25,1500

Priya,28,2200

Ravi,24,1800

Pandas can read this file and convert it into a DataFrame.

### Business Example

A company may receive daily sales data in a CSV file. The file can contain Product, Quantity, Price, and Sales columns. Pandas can read the file so the company can analyze its sales.

### AI/ML Use Case

In AI/ML projects, datasets such as customer data, house prices, or student records are often stored in CSV files. Pandas can read the CSV file into a DataFrame before data cleaning, feature preparation, and model training.

````python
# Import Pandas
import pandas as pd

# Create sales data
sales_data = {
    "Product": ["Laptop", "Phone", "Tablet"],
    "Quantity": [2, 5, 3],
    "Sales": [100000, 100000, 90000]
}

# Create a DataFrame
sales_df = pd.DataFrame(sales_data)

# Save the DataFrame as a CSV file
sales_df.to_csv("sales.csv", index=False)

print("CSV file created successfully.")
````

````python
# Import Pandas
import pandas as pd

# Read the CSV file
sales_df = pd.read_csv("sales.csv")

# Display the DataFrame
print(sales_df)
````

### Code Explanation

First, we import Pandas using `import pandas as pd`.

Next, `pd.read_csv("sales.csv")` reads the CSV file named `sales.csv`.

The data from the CSV file is stored in the `sales_df` DataFrame.

Finally, `print(sales_df)` displays the data.

### What the Code Does

The code reads tabular data from a CSV file and loads it into a Pandas DataFrame.

### Why the `pd.read_csv()` Function Is Used

The `pd.read_csv()` function is used to read CSV files into Pandas.

It allows us to work with CSV data using DataFrame operations such as filtering, cleaning, calculating, and analyzing.

### Expected Output

If `sales.csv` contains:

```text

Phone,5,100000
Tablet,3,90000

The CSV data has been loaded into a DataFrame, where the data is organized into rows and columns.

Where It Is Used in AI/ML or Data Analytics

# Topic 3 – JSON Files

## Step 1 – Understand the Concept

JSON stands for JavaScript Object Notation.

JSON is a structured data format commonly used to store and exchange data.

JSON data is usually organized using key-value pairs and can also contain lists and nested objects.

Commonly used Pandas functions:
- `pd.read_json()` – reads JSON data into a DataFrame.
- `df.to_json()` – writes a DataFrame to a JSON file.

JSON is commonly used in APIs, web applications, and data exchange between different systems.

## Step 2 – Demonstrate the Concept

### Real-World Analogy

Think of a student record containing labels such as Name, Age, and Marks with a value for each label. JSON stores information in a similar structured way using keys and values.

### Simple Example

A JSON file can contain:

{
    "Name": "Arun",
    "Age": 25,
    "Marks": 85
}

The keys describe the information, and the values contain the actual data.

### Business Example

An e-commerce application may receive customer or product information in JSON format from another system or API. Pandas can help convert this data into a DataFrame for analysis.

### AI/ML Use Case

AI/ML applications often receive data from APIs in JSON format. For example, an application may receive customer or product data from an API, convert it into a DataFrame, and use it for analysis or preprocessing.

````python
# Import Pandas
import pandas as pd

# Create customer data
customer_data = {
    "Name": ["Arun", "Priya", "Ravi"],
    "Age": [25, 28, 24],
    "Purchase": [1500, 2200, 1800]
}

# Create a DataFrame
customer_df = pd.DataFrame(customer_data)

# Save the DataFrame as a JSON file
customer_df.to_json("customers.json", orient="records")

print("JSON file created successfully.")
````

````python
# Read the JSON file
customer_df = pd.read_json("customers.json")

# Display the DataFrame
print(customer_df)
````

Then, we create customer data using a Python dictionary and convert it into a DataFrame.

The `to_json()` method saves the DataFrame as a JSON file.

The `orient="records"` parameter stores each DataFrame row as a separate JSON object.

Next, `pd.read_json()` reads the JSON file and loads the data into a DataFrame.

### What the Code Does

The code first creates customer data, saves it as a JSON file, and then reads the JSON file back into a Pandas DataFrame.

### Why the `to_json()` and `pd.read_json()` Functions Are Used

`to_json()` is used to write a DataFrame into JSON format.

`pd.read_json()` is used to read JSON data into a DataFrame.

The `orient="records"` option is useful because each row is stored as an individual record.

# Topic 5: Pickle Files

## Step 1: Understand the Concept

Pickle is a Python-specific format used to save and load Python objects, including Pandas DataFrames.

The `to_pickle()` method saves a DataFrame to a Pickle file. The `read_pickle()` function loads the saved object back into Python.

Pickle files preserve Python object structures and are useful when the same data needs to be reused later.

**Note:** Only load Pickle files from trusted sources because loading an untrusted file can execute malicious code.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Saving a DataFrame as a Pickle file is like storing your work in a saved file so you can reopen it later without creating everything again.

### Simple Example

A student saves marks in a DataFrame and loads the saved data the next day to continue working.

### Business Example

A company saves customer purchase data in a Pickle file and loads it later for sales analysis.

### AI/ML Use Case

An AI/ML developer saves a processed dataset and loads it again for model training or experimentation, avoiding repeated data preparation.

````python
# Import Pandas
import pandas as pd

# Create customer data
customer_data = {
    "Name": ["Arun", "Priya", "Ravi"],
    "Age": [25, 28, 24],
    "Purchase": [1500, 2200, 1800]
}

# Create a DataFrame
customer_df = pd.DataFrame(customer_data)

# Save the DataFrame as a Pickle file
customer_df.to_pickle("customers.pkl")

print("Pickle file created successfully.")

# Read the Pickle file
loaded_df = pd.read_pickle("customers.pkl")

# Display the loaded DataFrame
print(loaded_df)
````

## Step 3: Implement the Concept

### Code Explanation

- `import pandas as pd` imports Pandas.
- `customer_data` stores customer information in a dictionary.
- `pd.DataFrame(customer_data)` converts the dictionary into a DataFrame.
- `customer_df.to_pickle("customers.pkl")` saves the DataFrame in a Pickle file.
- `pd.read_pickle("customers.pkl")` loads the saved DataFrame.
- `print(loaded_df)` displays the loaded data.

### What the Code Does

The code creates a customer DataFrame, saves it as `customers.pkl`, and loads the saved data back into Python.

### Why the Function/Method Is Used

- `pd.DataFrame()` creates a table from the dictionary.
- `to_pickle()` saves the DataFrame while preserving its Python object structure.
- `read_pickle()` loads the saved object for reuse.
- `print()` displays the result.

### Expected Output

```text
Pickle file created successfully.
    Name  Age  Purchase
0   Arun   25      1500
1  Priya   28      2200
2   Ravi   24      1800
```

The loaded DataFrame contains the same customer records that were saved in the Pickle file.

### Where It Is Used in AI/ML or Data Analytics

Pickle can be used to save and reload processed DataFrames and other Python objects during data analysis and machine learning workflows. Only use trusted Pickle files.

# Topic 6: SQL Tables

## Step 1: Understand the Concept

SQL tables store structured data in rows and columns inside a database.

Pandas provides `read_sql()` to read SQL query results into a DataFrame and `to_sql()` to write DataFrame data into a database table.

SQLite is a lightweight database that can be used directly from Python.

These functions are useful for transferring data between databases and Pandas.

**Important:** Use trusted database connections and appropriate permissions when reading or writing data.

## Step 2: Demonstrate the Concept

### Real-World Analogy

A database is like a digital register that stores information in organized tables. Pandas helps us bring that information into a DataFrame for analysis.

### Simple Example

A student database stores student names and marks. Python reads the student table into a DataFrame to calculate or compare marks.

### Business Example

A company stores product sales in an SQL table. A data analyst reads the sales records into Pandas to calculate total sales and identify popular products.

### AI/ML Use Case

An AI/ML developer reads customer purchase records from a database into a DataFrame and uses the data for cleaning, feature preparation, and model training.

````python
# Import Pandas and SQLite
import pandas as pd
import sqlite3

# Create sales data
sales_data = {
    "Product": ["Laptop", "Phone", "Tablet"],
    "Quantity": [2, 5, 3],
    "Sales": [100000, 100000, 90000]
}

# Create a DataFrame
sales_df = pd.DataFrame(sales_data)

# Connect to an SQLite database
connection = sqlite3.connect("sales.db")

# Write the DataFrame to an SQL table
sales_df.to_sql("sales", connection, if_exists="replace", index=False)

# Read the SQL table into a DataFrame
loaded_df = pd.read_sql("SELECT * FROM sales", connection)

# Display the loaded data
print(loaded_df)

# Close the database connection
connection.close()
````

## Step 3: Implement the Concept

### Code Explanation

- `import pandas as pd` imports Pandas for DataFrame operations.
- `import sqlite3` imports Python's built-in SQLite database module.
- `pd.DataFrame(sales_data)` creates a DataFrame from the sales dictionary.
- `sqlite3.connect("sales.db")` connects to an SQLite database file. If the file does not exist, SQLite creates it.
- `sales_df.to_sql()` saves the DataFrame in a table named `sales`.
- `if_exists="replace"` replaces the table if it already exists.
- `index=False` prevents the DataFrame index from being saved as an extra column.
- `pd.read_sql("SELECT * FROM sales", connection)` reads all rows and columns from the SQL table.
- `connection.close()` closes the database connection.

### What the Code Does

The code creates sales data, saves it in an SQLite database table, reads the table back into Pandas, and displays the result.

### Why the Function/Method Is Used

- `sqlite3.connect()` opens a database connection.
- `to_sql()` transfers DataFrame data into an SQL table.
- `read_sql()` retrieves SQL query results as a DataFrame.
- `SELECT * FROM sales` selects all columns and rows from the `sales` table.
- `close()` releases the database connection.

### Expected Output

```text
   Product  Quantity   Sales
0   Laptop         2  100000
1    Phone         5  100000
2   Tablet         3   90000
```

The output shows that the data was written to the SQL table and successfully read back into a DataFrame.

### Where It Is Used in AI/ML or Data Analytics

SQL tables are commonly used to store customer, product, transaction, and sales data. Pandas can retrieve this data for analysis, preprocessing, and preparing features for machine learning models.

# Topic 7: Reading Multiple Files

## Step 1: Understand the Concept

Reading multiple files means loading data from several files into Python for analysis.

The `glob` module finds files that match a filename pattern. Pandas `read_csv()` reads each CSV file, and `concat()` combines multiple DataFrames into one.

This approach is useful when data is stored in separate files by month, department, or location.

The files should have compatible column names and data structures when combining them.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine collecting attendance sheets from several classrooms. Instead of checking each sheet separately, you collect them and combine the records for one report.

### Simple Example

A student has marks saved in separate files for different subjects. Python can read the files and combine the marks into one DataFrame.

### Business Example

A company saves sales records in separate CSV files for January, February, and March. An analyst combines them to calculate quarterly sales.

### AI/ML Use Case

An AI/ML developer combines datasets collected from different sources or time periods to prepare a larger dataset for analysis and model training.

````python
# Import Pandas and glob
import pandas as pd
import glob

# Create sample sales data
january_data = {
    "Product": ["Laptop", "Phone"],
    "Sales": [50000, 20000]
}

february_data = {
    "Product": ["Tablet", "Headphones"],
    "Sales": [30000, 5000]
}

# Save the sample data as separate CSV files
pd.DataFrame(january_data).to_csv("sales_january.csv", index=False)
pd.DataFrame(february_data).to_csv("sales_february.csv", index=False)

# Find all sales CSV files
file_list = glob.glob("sales_*.csv")

# Read each file into a DataFrame
dataframes = [pd.read_csv(file) for file in file_list]

# Combine all DataFrames
combined_df = pd.concat(dataframes, ignore_index=True)

# Display the combined data
print(combined_df)
````

## Step 3: Implement the Concept

### Code Explanation

- `import pandas as pd` imports Pandas.
- `import glob` imports the module used to find matching filenames.
- `january_data` and `february_data` store sample sales records.
- `to_csv()` saves each dictionary as a separate CSV file.
- `glob.glob("sales_*.csv")` finds CSV filenames that start with `sales_`.
- `[pd.read_csv(file) for file in file_list]` reads every matching file into a DataFrame.
- `pd.concat(dataframes, ignore_index=True)` combines all DataFrames row by row and creates a fresh sequential index.
- `print(combined_df)` displays the combined records.

### What the Code Does

The code creates two sample CSV files, finds both files, reads them into Pandas, and combines their records into one DataFrame.

### Why the Function/Method Is Used

- `glob.glob()` finds multiple files using a filename pattern.
- `pd.read_csv()` reads data from each CSV file.
- `pd.concat()` combines DataFrames.
- `ignore_index=True` creates a new index for the combined data instead of retaining the original indexes.

### Expected Output

```text id="8w2n1x"
      Product  Sales
0      Laptop  50000
1       Phone  20000
2      Tablet  30000
3  Headphones   5000
```

The output contains records from both CSV files in a single DataFrame. The order can depend on the order in which the files are returned.

### Where It Is Used in AI/ML or Data Analytics

Reading multiple files is useful for combining monthly sales reports, customer records, or datasets collected from different sources before data cleaning, analysis, and machine learning.

# Topic 8: File Path Handling

## Step 1: Understand the Concept

File path handling means specifying and managing the locations of files and folders in Python.

A relative path refers to a location from the current working directory. An absolute path specifies the complete file location.

The `os.getcwd()` function returns the current working directory, `os.path.exists()` checks whether a path exists, and `os.path.join()` combines folder and file names.

Correct file paths help prevent `FileNotFoundError` when reading or writing files.

## Step 2: Demonstrate the Concept

### Real-World Analogy

A file path is like a home address. If the address is correct, you can find the right house. Similarly, Python needs the correct path to find a file.

### Simple Example

A student saves a marks file inside a folder named `Data`. Python uses the folder path and filename to locate the file.

### Business Example

A company stores monthly sales reports in separate folders. Correct path handling allows an analyst to read the required report without searching manually.

### AI/ML Use Case

An AI/ML developer uses file paths to locate datasets, configuration files, and saved model files during data preparation and model development.

````python
# Import Pandas and os
import pandas as pd
import os

# Create sample sales data
sales_data = {
    "Product": ["Laptop", "Phone", "Tablet"],
    "Sales": [50000, 20000, 30000]
}

# Create a DataFrame
sales_df = pd.DataFrame(sales_data)

# Create a folder for the file
os.makedirs("Data", exist_ok=True)

# Build the file path
file_path = os.path.join("Data", "sales.csv")

# Save the DataFrame
sales_df.to_csv(file_path, index=False)

# Display the current working directory
print("Current directory:", os.getcwd())

# Check whether the file exists
print("File exists:", os.path.exists(file_path))

# Read the CSV file using its path
loaded_df = pd.read_csv(file_path)

# Display the loaded data
print(loaded_df)
````

## Step 3: Implement the Concept

### Code Explanation

- `import pandas as pd` imports Pandas for data handling.
- `import os` imports tools for working with folders and file paths.
- `pd.DataFrame(sales_data)` creates a DataFrame.
- `os.makedirs("Data", exist_ok=True)` creates the `Data` folder if it does not already exist.
- `os.path.join("Data", "sales.csv")` combines the folder and filename into a file path.
- `sales_df.to_csv(file_path, index=False)` saves the DataFrame at that location.
- `os.getcwd()` displays the current working directory.
- `os.path.exists(file_path)` checks whether the file exists.
- `pd.read_csv(file_path)` reads the CSV file using the specified path.

### What the Code Does

The code creates a DataFrame, creates a folder, saves the DataFrame as a CSV file inside that folder, checks the file location, and reads the saved data back into Pandas.

### Why the Function/Method Is Used

- `os.makedirs()` creates a folder.
- `os.path.join()` builds a path using the correct separator for the operating system.
- `os.getcwd()` helps identify where relative paths start.
- `os.path.exists()` checks whether a file or folder exists before using it.
- `to_csv()` saves tabular data.
- `read_csv()` loads the saved data into a DataFrame.

### Expected Output

```text
Current directory: [Your current working directory]
File exists: True
   Product  Sales
0   Laptop  50000
1    Phone  20000
2   Tablet  30000
```

The actual directory path depends on where your Jupyter Notebook is running. `File exists: True` confirms that the CSV file was saved at the specified path.

### Where It Is Used in AI/ML or Data Analytics

File path handling is used to locate datasets, load training and testing data, save analysis results, and access configuration or model files reliably.
