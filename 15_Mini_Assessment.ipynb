## Question 6 – Merge Two DataFrames

### Coding Exercise:

Write a Python program using Pandas to create two DataFrames.

**DataFrame 1 – Customers**

| Customer_ID | Name |
|---|---|
| 101 | Ravi |
| 102 | Priya |
| 103 | Arun |

**DataFrame 2 – Orders**

| Customer_ID | Amount |
|---|---:|
| 101 | 2500 |
| 102 | 3500 |
| 103 | 2000 |

Merge both DataFrames using `Customer_ID` and display the merged DataFrame.

```python
import pandas as pd

customers = pd.DataFrame({
    "Customer_ID": [101, 102, 103],
    "Name": ["Ravi", "Priya", "Arun"]
})

orders = pd.DataFrame({
    "Customer_ID": [101, 102, 103],
    "Amount": [2500, 3500, 2000]
})

merged_df = pd.merge(customers, orders, on="Customer_ID")

print(merged_df)
```

## Question 7 – GroupBy and Aggregation

### Coding Exercise:

Write a Python program using Pandas to create a DataFrame with the following data:

| Product | Category | Sales |
|---|---|---:|
| Laptop | Electronics | 50000 |
| Phone | Electronics | 20000 |
| Shirt | Clothing | 1500 |
| Jeans | Clothing | 2500 |

Perform the following tasks:

1. Group the data by `Category`.
2. Calculate the total Sales for each category.
3. Calculate the average Sales for each category.
4. Display the results.

```python
import pandas as pd

sales_df = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Shirt", "Jeans"],
    "Category": ["Electronics", "Electronics", "Clothing", "Clothing"],
    "Sales": [50000, 20000, 1500, 2500]
})

result = sales_df.groupby("Category")["Sales"].agg(
    Total_Sales="sum",
    Average_Sales="mean"
)

print(result)
```

## Question 8 – Filter Rows Using Multiple Conditions

### Coding Exercise:

Write a Python program using Pandas to create a DataFrame with the following data:

| Name | Age | Salary |
|---|---:|---:|
| Ravi | 24 | 25000 |
| Priya | 27 | 35000 |
| Arun | 22 | 22000 |
| Meena | 30 | 45000 |
| Karthik | 28 | 40000 |

Filter and display employees whose **Age is greater than 25 AND Salary is greater than 30000**.

```python
import pandas as pd

employee_df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun", "Meena", "Karthik"],
    "Age": [24, 27, 22, 30, 28],
    "Salary": [25000, 35000, 22000, 45000, 40000]
})

filtered_df = employee_df[
    (employee_df["Age"] > 25) &
    (employee_df["Salary"] > 30000)
]

print(filtered_df)
```

## Question 9 – Handle Missing Values

### Coding Exercise:

Write a Python program using Pandas to create a DataFrame with the following data:

| Name | Age | Salary |
|---|---:|---:|
| Ravi | 24 | 25000 |
| Priya | NaN | 35000 |
| Arun | 22 | NaN |
| Meena | 30 | 45000 |

Perform the following tasks:

1. Identify missing values.
2. Fill missing Age values using the median.
3. Fill missing Salary values using the mean.
4. Display the updated DataFrame.

```python
import pandas as pd

employee_df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun", "Meena"],
    "Age": [24, None, 22, 30],
    "Salary": [25000, 35000, None, 45000]
})

print("Missing Values:")
print(employee_df.isnull())

employee_df["Age"] = employee_df["Age"].fillna(
    employee_df["Age"].median()
)

employee_df["Salary"] = employee_df["Salary"].fillna(
    employee_df["Salary"].mean()
)

print("\nUpdated DataFrame:")
print(employee_df)
```

## Question 10 – Calculate Statistical Measures Using NumPy

### Coding Exercise:

Write a Python program using NumPy to calculate the following statistical measures for the dataset:

`[10, 20, 30, 40, 100]`

1. Mean
2. Median
3. Variance
4. Standard deviation
5. Minimum value
6. Maximum value

```python
import numpy as np

data = np.array([10, 20, 30, 40, 100])

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Variance:", np.var(data))
print("Standard Deviation:", np.std(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
```

## Question 11 – Optimize a DataFrame for Memory Usage

### Coding Exercise:

Write a Python program using Pandas to:

1. Create a DataFrame with an `Age` column containing `[20, 25, 30, 35, 40, 45, 50]`.
2. Display the original data type and memory usage.
3. Convert the `Age` column to a suitable smaller integer data type.
4. Display the updated data type and memory usage.
5. Compare the memory usage before and after optimization.

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40, 45, 50]
})

print("Original Data Type:", df["Age"].dtype)
print("Original Memory Usage:", df["Age"].memory_usage(deep=True), "bytes")

df["Age"] = df["Age"].astype("int8")

print("\nOptimized Data Type:", df["Age"].dtype)
print("Optimized Memory Usage:", df["Age"].memory_usage(deep=True), "bytes")
```
