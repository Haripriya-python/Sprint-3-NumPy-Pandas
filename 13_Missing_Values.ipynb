# Missing Values - isnull()

## Step 1: Understand the Concept

`isnull()` is used to identify missing values in a Series or DataFrame.

It checks each value and returns:

- `True` if the value is missing.
- `False` if the value is available.

Missing values are commonly represented as `NaN` in Pandas.

Why do we use `isnull()`?

- To identify missing values.
- To understand the quality of a dataset.
- To find columns containing missing data.
- To decide how missing values should be handled before analysis or machine learning.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a customer registration form where some customers did not provide their age.

`isnull()` works like checking the form and marking which age fields were left empty.

### Simple Example

Suppose we have:

| Name | Age |
|---|---:|
| Ravi | 25 |
| Priya | NaN |
| Arun | 30 |

When we use `isnull()` on the `Age` column:

- Ravi → `False`
- Priya → `True`
- Arun → `False`

### Business Example

A company may have a customer dataset where some customers have not provided their phone number, age, or email address.

`isnull()` can help the company identify which customer records contain missing information.

### AI/ML Use Case

Before training a machine learning model, we need to identify missing values in the dataset.

`isnull()` helps detect missing data so that we can decide whether to remove, replace, or estimate those values.

```python
import pandas as pd

customer_df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun"],
    "Age": [25, None, 30],
    "City": ["Chennai", "Madurai", None]
})

print(customer_df)
print("\nMissing Values:")
print(customer_df.isnull())
```

## Step 3: Code Explanation

### Code Explanation

`pd.DataFrame()` creates a customer dataset with Name, Age, and City columns.

`None` is used to represent missing values in the dataset.

`isnull()` checks every value in the DataFrame and identifies whether it is missing.

### What the Code Does

The code creates a DataFrame and checks all columns for missing values.

It returns `True` for missing values and `False` for values that are present.

### Why the Function/Method Is Used

- `pd.DataFrame()` → Creates the dataset.
- `isnull()` → Identifies missing values.
- `None` → Represents a missing value in the example dataset.

### Expected Output

The result will contain:

- `True` for missing values.
- `False` for available values.

For example:

- Priya's Age → `True`
- Arun's Age → `False`
- Arun's City → `True`

This helps us quickly identify where missing data exists.

### Where It Is Used in AI/ML or Data Analytics

`isnull()` is commonly used during data cleaning and exploratory data analysis to identify missing values before performing analysis or training machine learning models.

# Missing Values - notnull()

## Step 1: Understand the Concept

`notnull()` is used to identify values that are not missing in a Series or DataFrame.

It checks each value and returns:

- `True` if the value is available.
- `False` if the value is missing.

`notnull()` is the opposite of `isnull()`.

Why do we use `notnull()`?

- To identify available values.
- To filter rows containing valid data.
- To check whether required information is present.
- To work with complete records during data analysis.

    ## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine checking customer registration forms.

If a customer has provided their phone number, we mark it as available.
If they have left it empty, we mark it as missing.

`notnull()` performs this check automatically.

### Simple Example

Suppose we have:

| Name | Age |
|---|---:|
| Ravi | 25 |
| Priya | NaN |
| Arun | 30 |

Using `notnull()` on the Age column gives:

- Ravi → `True`
- Priya → `False`
- Arun → `True`

### Business Example

A company may need to contact customers using their phone numbers.

`notnull()` can help identify customer records where the phone number is available.

### AI/ML Use Case

During data preprocessing, `notnull()` can be used to identify records that contain required values before further analysis or model training.

```python
import pandas as pd

customer_df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun"],
    "Age": [25, None, 30],
    "City": ["Chennai", "Madurai", None]
})

print(customer_df)
print("\nAvailable Values:")
print(customer_df.notnull())
```

## Step 3: Code Explanation

### Code Explanation

`pd.DataFrame()` creates a customer dataset containing Name, Age, and City.

`None` represents a missing value.

`notnull()` checks every value and identifies whether the value is available.

### What the Code Does

The code checks all values in the DataFrame.

It returns `True` when a value is available and `False` when a value is missing.

### Why the Function/Method Is Used

- `pd.DataFrame()` → Creates the dataset.
- `notnull()` → Identifies values that are not missing.
- `None` → Represents a missing value.

### Expected Output

The result will contain:

- `True` for available values.
- `False` for missing values.

For example:

- Ravi's Age → `True`
- Priya's Age → `False`
- Arun's Age → `True`
- Arun's City → `False`

### Where It Is Used in AI/ML or Data Analytics

`notnull()` is useful during data cleaning to identify records with available information and to filter datasets based on the presence of required values.

# Missing Values - dropna()

## Step 1: Understand the Concept

`dropna()` is used to remove rows or columns that contain missing values.

By default, `dropna()` removes rows containing at least one missing value.

Important parameters:

- `axis=0` → Removes rows. This is the default.
- `axis=1` → Removes columns.
- `inplace=True` → Applies the change directly to the original DataFrame.

Why do we use `dropna()`?

- To remove incomplete records.
- To create a cleaner dataset.
- To remove rows when missing information is not useful.
- To prepare data for analysis or machine learning.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a customer registration form where some required information is missing.

If a company decides that incomplete forms cannot be used, those incomplete forms can be removed.

`dropna()` performs this removal automatically.

### Simple Example

Suppose we have:

| Name | Age |
|---|---:|
| Ravi | 25 |
| Priya | NaN |
| Arun | 30 |

When `dropna()` is applied, the row containing the missing Age value is removed.

### Business Example

A company may have sales records where some rows are missing important information such as product name or sales amount.

If those incomplete records cannot be used, `dropna()` can remove them.

### AI/ML Use Case

In machine learning preprocessing, rows containing missing values may sometimes be removed when only a small number of records are incomplete and removing them will not significantly affect the dataset.

```python
import pandas as pd

customer_df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun", "Meena"],
    "Age": [25, None, 30, None],
    "City": ["Chennai", "Madurai", None, "Coimbatore"]
})

print("Original DataFrame:")
print(customer_df)

clean_df = customer_df.dropna()

print("\nDataFrame After dropna():")
print(clean_df)
```

## Step 3: Code Explanation

### Code Explanation

`pd.DataFrame()` creates a customer dataset containing Name, Age, and City.

Some values are missing and are represented using `None`.

`dropna()` removes rows that contain at least one missing value.

### What the Code Does

The original DataFrame contains four customer records.

After applying `dropna()`, only rows with complete information remain.

In this example:

- Ravi has complete data, so the row remains.
- Priya has a missing Age, so the row is removed.
- Arun has a missing City, so the row is removed.
- Meena has a missing Age, so the row is removed.

### Why the Function/Method Is Used

- `pd.DataFrame()` → Creates the dataset.
- `dropna()` → Removes rows containing missing values.
- `clean_df` → Stores the cleaned DataFrame without changing the original DataFrame.

### Expected Output

The original DataFrame contains all four rows.

After `dropna()`, only Ravi's complete record remains.

The output will be similar to:

       Name   Age     City
0      Ravi  25.0  Chennai

### Where It Is Used in AI/ML or Data Analytics

`dropna()` is used during data cleaning when incomplete records should be removed before analysis or machine learning.

It should be used carefully because removing too many rows can reduce the amount of available data.

# Missing Values - fillna()

## Step 1: Understand the Concept

`fillna()` is used to replace missing values with a specified value.

Instead of removing rows that contain missing values, we can fill the missing values with a suitable value.

Common ways to use `fillna()`:

- Fill with a fixed value.
- Fill with the mean.
- Fill with the median.
- Fill with the mode.

Why do we use `fillna()`?

- To keep useful records in the dataset.
- To replace missing values.
- To prepare complete data for analysis.
- To avoid losing rows unnecessarily.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a customer form where the Age field is empty.

Instead of throwing away the entire form, we can estimate or provide a suitable value for the missing Age.

`fillna()` works in a similar way by replacing missing values.

### Simple Example

Suppose we have:

| Name | Age |
|---|---:|
| Ravi | 25 |
| Priya | NaN |
| Arun | 30 |

If we use `fillna(27)`, the missing Age is replaced with 27.

### Business Example

A company may have customer data where some customers have missing ages.

Instead of deleting those customer records, the company can fill the missing ages using an appropriate statistical value such as the median.

### AI/ML Use Case

Machine learning algorithms often require complete input data.

`fillna()` can be used during preprocessing to replace missing values using methods such as mean, median, or mode.

```python
import pandas as pd

customer_df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun", "Meena"],
    "Age": [25, None, 30, None]
})

print("Original DataFrame:")
print(customer_df)

customer_df["Age"] = customer_df["Age"].fillna(
    customer_df["Age"].mean()
)

print("\nDataFrame After Filling Missing Ages:")
print(customer_df)
```

## Step 3: Code Explanation

### Code Explanation

`pd.DataFrame()` creates a customer dataset with some missing Age values.

`mean()` calculates the average of the available Age values.

`fillna()` replaces the missing Age values with that calculated mean.

### What the Code Does

The available ages are:

- 25
- 30

The mean is:

27.5

Therefore, the missing Age values are replaced with 27.5.

### Why the Function/Method Is Used

- `pd.DataFrame()` → Creates the dataset.
- `mean()` → Calculates the average of available values.
- `fillna()` → Replaces missing values.
- `customer_df["Age"]` → Selects the Age column.

### Expected Output

The output will be similar to:

      Name   Age
0     Ravi  25.0
1    Priya  27.5
2     Arun  30.0
3    Meena  27.5

The missing Age values have been replaced with the mean age of 27.5.

### Where It Is Used in AI/ML or Data Analytics

`fillna()` is commonly used during data preprocessing to handle missing values while preserving useful records for analysis and machine learning.

# Missing Values - interpolate()

## Step 1: Understand the Concept

`interpolate()` is used to estimate and fill missing values based on surrounding values.

It is especially useful when data has a natural order, such as time-series data.

For example:

- Monday → 100
- Tuesday → Missing
- Wednesday → 140

Interpolation can estimate Tuesday as 120 because it is between 100 and 140.

Why do we use `interpolate()`?

- To estimate missing values.
- To preserve the pattern of ordered data.
- To handle missing values in time-series data.
- To avoid simply replacing missing values with one fixed value.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a car traveling from one place to another.

At 10:00 AM, the car is at 100 km.
At 12:00 PM, the car is at 140 km.

If the location at 11:00 AM is missing, we can estimate that the car was around 120 km.

This is similar to interpolation.

### Simple Example

Suppose daily sales are:

| Day | Sales |
|---|---:|
| Monday | 100 |
| Tuesday | NaN |
| Wednesday | 140 |

The missing Tuesday value can be estimated as 120 because it lies between 100 and 140.

### Business Example

A company may record daily sales or website traffic.

If one day's value is missing, interpolation can estimate the missing value using nearby observations.

### AI/ML Use Case

Interpolation can be useful during data preprocessing for time-series and sensor datasets where values follow an ordered pattern.

Examples include:

- Temperature readings
- Sensor data
- Stock data
- Daily sales
- Website traffic

import pandas as pd

sales_df = pd.DataFrame({
    "Day": ["Monday", "Tuesday", "Wednesday"],
    "Sales": [100, None, 140]
})

print("Original DataFrame:")
print(sales_df)

sales_df["Sales"] = sales_df["Sales"].interpolate()

print("\nDataFrame After Interpolation:")
print(sales_df)

## Step 3: Code Explanation

### Code Explanation

`pd.DataFrame()` creates a sales dataset.

The Sales value for Tuesday is missing.

`.interpolate()` estimates the missing value using the surrounding values.

The available values are:

- Monday → 100
- Wednesday → 140

The estimated middle value is 120.

### What the Code Does

The code identifies the missing Sales value and fills it using interpolation.

The missing Tuesday Sales value becomes 120.

### Why the Function/Method Is Used

- `pd.DataFrame()` → Creates the sales dataset.
- `.interpolate()` → Estimates and fills missing values based on surrounding values.
- `sales_df["Sales"]` → Selects the Sales column.

### Expected Output

The output will be similar to:

        Day  Sales
0    Monday  100.0
1   Tuesday  120.0
2 Wednesday  140.0

The missing Tuesday value is estimated as 120 based on the values before and after it.

### Where It Is Used in AI/ML or Data Analytics

`interpolate()` is useful for handling missing values in ordered and time-series datasets such as sales, sensor readings, temperature measurements, stock prices, and website traffic.

# Missing Value Handling Strategies

## Step 1: Understand the Concept

Missing Value Handling Strategies are methods used to manage missing data in a dataset.

There is no single strategy that works for every dataset.

The appropriate strategy depends on:

- The type of data.
- The number of missing values.
- The distribution of the data.
- The importance of the missing records.
- The business requirement.

Common strategies include:

1. Removing missing records using `dropna()`.
2. Filling numerical values using the mean.
3. Filling numerical values using the median.
4. Filling categorical values using the mode.
5. Estimating ordered values using `interpolate()`.
6. Using forward fill or backward fill when the previous or next value is meaningful.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a customer form with some empty fields.

We have different choices:

- If the missing information is not important, we can remove the record.
- If the value is numerical, we can use a suitable statistical value.
- If the value is categorical, we can use the most common category.
- If the data follows a time-based pattern, we can estimate the missing value from nearby values.

The correct choice depends on the situation.

### Simple Example

Suppose we have customer age data:

| Age |
|---:|
| 20 |
| 25 |
| NaN |
| 30 |
| 100 |

The value 100 is an outlier.

Using the median may be more suitable than the mean because the median is less affected by extreme values.

### Business Example

A retail company may have missing sales values.

Possible strategies include:

- Remove records if only a very small number are missing.
- Use the median for numerical customer information when appropriate.
- Use the mode for categorical information such as city or product category.
- Use interpolation for ordered daily sales data.

### AI/ML Use Case

Missing value handling is an important part of data preprocessing before machine learning.

Choosing the right strategy helps preserve useful information and prevents poor-quality data from affecting model training and analysis.

```python
import pandas as pd

customer_df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun", "Meena", "Karthik"],
    "Age": [25, None, 30, 28, None],
    "City": ["Chennai", "Madurai", None, "Chennai", "Madurai"]
})

print("Original DataFrame:")
print(customer_df)

# Fill numerical missing values with the median
customer_df["Age"] = customer_df["Age"].fillna(
    customer_df["Age"].median()
)

# Fill categorical missing values with the mode
customer_df["City"] = customer_df["City"].fillna(
    customer_df["City"].mode()[0]
)

print("\nDataFrame After Handling Missing Values:")
print(customer_df)
```

## Step 3: Code Explanation

### Code Explanation

The DataFrame contains missing values in the Age and City columns.

For the Age column, the median is used to replace missing values.

For the City column, the mode is used to replace the missing value.

### What the Code Does

The code handles missing values using different strategies based on the type of data.

For Age:

- The median of the available ages is calculated.
- Missing Age values are replaced with the median.

For City:

- The most frequently occurring city is found using `mode()`.
- The missing City value is replaced with that city.

### Why the Function/Method Is Used

- `median()` → Finds the middle value of numerical data.
- `mode()` → Finds the most frequently occurring value.
- `fillna()` → Replaces missing values.
- `mode()[0]` → Selects the first mode value for filling the missing entry.

### Expected Output

The missing Age values will be replaced with the median age.

The missing City value will be replaced with the most frequently occurring city.

The resulting DataFrame will contain no missing values in the Age and City columns.

### Where It Is Used in AI/ML or Data Analytics

Missing value handling is used during data preprocessing to create cleaner and more complete datasets for data analysis and machine learning.

The strategy should be selected based on the data type, distribution, missing-value pattern, and business requirement.
