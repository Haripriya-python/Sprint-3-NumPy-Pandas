# 1. What is Pandas?

## Step 1 – Understand the Concept

Pandas is a Python library used for working with structured and tabular data.

It provides easy-to-use tools for storing, organizing, cleaning, manipulating, and analyzing data.

Pandas is commonly used with data such as:

- Customer records
- Sales information
- Employee details
- Student marks
- Product information
- Transaction data

Pandas mainly provides two important data structures:

- **Series** → One-dimensional labeled data
- **DataFrame** → Two-dimensional tabular data with rows and columns

### Importing Pandas

Pandas is commonly imported using:

`import pandas as pd`

Here, `pd` is an alias for Pandas. It allows us to access Pandas functions and classes using `pd`.

### Example

A table containing customer names, ages, and purchase amounts can be stored and analyzed using a Pandas DataFrame.

## Step 2 – Demonstrate the Concept

### Real-World Analogy

Think about an Excel spreadsheet containing customer information.

The spreadsheet has rows representing individual customers and columns representing details such as name, age, city, and purchase amount.

Pandas provides similar table-based data handling capabilities directly inside Python.

### Simple Example

Suppose we have customer data:

| Name | Age | Purchase |
|------|-----|----------|
| Arun | 25 | 1500 |
| Priya | 28 | 2200 |
| Ravi | 24 | 1800 |

Instead of manually working with this data, Pandas can store it in a structured format and allow us to analyze it using Python.

### Business Example

An e-commerce company may have thousands of customer transactions.

Using Pandas, the company can organize the transaction data and calculate values such as total sales, average purchase amount, and number of customers.

### AI/ML Use Case

Pandas is widely used during the data preparation stage of AI/ML projects.

For example, customer data can be loaded into a DataFrame, missing values can be handled, columns can be selected, and the data can be prepared before passing it to a machine learning model.

````python
# Import Pandas
import pandas as pd

# Create customer data
customer_data = {
    "Name": ["Arun", "Priya", "Ravi"],
    "Age": [25, 28, 24],
    "Purchase": [1500, 2200, 1800]
}

# Create a DataFrame
customer_df = pd.DataFrame(customer_data)

# Display the DataFrame
print(customer_df)
````

## Step 3 – Explanation

### Code Explanation

- `import pandas as pd` imports the Pandas library and gives it the alias `pd`.
- `customer_data` is a Python dictionary containing customer information.
- `"Name"`, `"Age"`, and `"Purchase"` are the column names.
- The lists contain the corresponding values for each customer.
- `pd.DataFrame(customer_data)` converts the dictionary into a Pandas DataFrame.
- `customer_df` stores the created DataFrame.
- `print(customer_df)` displays the table.

### What the Code Does

The code creates simple customer information using a Python dictionary and converts it into a table using Pandas.

Each dictionary key becomes a column, and the values become the data inside that column.

### Why `import pandas as pd` Is Used

We import Pandas because it provides functions and data structures for working with tabular data.

`pd` is a commonly used alias that makes the code shorter and easier to read.

### Why `pd.DataFrame()` Is Used

`pd.DataFrame()` is used to create a DataFrame from structured data.

A DataFrame represents data in rows and columns, similar to an Excel table.

### Expected Output

The output contains three customer records with three columns:

```text
    Name  Age  Purchase
0   Arun   25      1500
1  Priya   28      2200
2   Ravi   24      1800

The numbers on the left (0, 1, 2) are the default row indexes.

In AI/ML, Pandas can be used to load datasets, inspect data, clean missing values, select features, transform data, and prepare datasets before machine learning model training.

# 3. Series

## Step 1 – Understand the Concept

A Pandas Series is a one-dimensional labeled array capable of holding data of any type (integer, string, float, python objects, etc.).

It is similar to a single column in an Excel sheet or a database table.

A Series consists of two main parts:
- **Values** → The actual data elements inside the Series.
- **Index** → The row labels associated with each data element.

### Default vs Custom Indexing
- By default, Pandas assigns integer indexes starting from `0, 1, 2...`
- We can also assign custom string labels as indexes (for example: fruit names, product IDs, or days).

### Difference Between List and Series
- A standard **Python List** only has hidden positional numbers (`0, 1, 2`).
- A **Pandas Series** has labeled indexes and built-in methods for fast statistical operations.

## Step 2 – Demonstrate the Concept

### Real-World Analogy
Think of a price tag attached to products in a supermarket shelf.

The product names act as the **Index** (labels) and the prices act as the **Values**.

### Simple Example
Suppose we have fruit prices:

| Index (Fruit) | Value (Price) |
|---------------|---------------|
| Apple         | 100           |
| Banana        | 20            |
| Mango         | 150           |

Here, `Apple`, `Banana`, and `Mango` are row labels, and `100`, `20`, `150` are the values.

### Business Example
In an e-commerce inventory system, a Series can track product stock levels:
- Index: Product SKUs (`SKU_1`, `SKU_2`, `SKU_3`)
- Values: Available Stock Quantity (`50`, `0`, `120`)

The store manager can instantly query `stock['SKU_2']` to check stock levels without scanning the entire database.

### AI/ML Use Case
In machine learning tasks, a single target variable column (like predicting `House Price` or `Customer Churn`) is extracted as a Pandas Series and passed to the ML model as the label vector $y$.

````python
# Import Pandas
import pandas as pd

# Create data list and index list
prices = [100, 20, 150]
fruits = ["Apple", "Banana", "Mango"]

# Create a Pandas Series with custom index
fruit_series = pd.Series(data=prices, index=fruits)

# Display the Series
print(fruit_series)

# Accessing a single element using label index
print("\nPrice of Banana:")
print(fruit_series["Banana"])
````

## Step 3 – Explanation

### Code Explanation
- `import pandas as pd` imports the Pandas library into Python.
- `prices = [100, 20, 150]` is a standard list containing price numbers.
- `fruits = ["Apple", "Banana", "Mango"]` is a list containing string labels for the index.
- `pd.Series(data=prices, index=fruits)` creates a Pandas Series mapping the prices to the corresponding fruit labels.
- `fruit_series` stores the created Series object.
- `print(fruit_series["Banana"])` accesses and prints the value associated with the label `"Banana"`.

### What the Code Does
The code combines two simple Python lists into a 1-dimensional labeled array (Series) and fetches a value using its assigned label.

### Why `pd.Series()` Is Used
`pd.Series()` is used to store and manipulate a single sequence of data elements with labeled indexing, making data lookup clear and simple.

### Expected Output
```text
Apple     100
Banana     20
Mango     150
dtype: int64

Price of Banana:
20

### Where It Is Used in AI/ML or Data Analytics
Target Variable Isolation: In Machine Learning pipelines, target columns (labels) are extracted as Pandas Series before model training.

Feature Extraction: Operations on single metrics (e.g., extracting transaction timestamps or calculating mean age) are performed directly on Pandas Series objects.

Data Filtering: Series objects are frequently used to create conditional boolean masks for filtering tables in Analytics.

# 4.DataFrame

## Step 1 – Understand the Concept

A Pandas DataFrame is a two-dimensional labeled data structure with columns of potentially different types.

It is equivalent to a spreadsheet, an Excel table, or a database SQL table.

A DataFrame consists of three main components:
- **Data** → The actual values stored in the table.
- **Rows (Index)** → Horizontal labels that identify individual records.
- **Columns** → Vertical labels that identify specific features or attributes.

### Key Characteristics of a DataFrame
- Contains both row and column labels.
- Can store mixed data types (integers, strings, floats, booleans) in different columns.
- Allows easy mathematical operations, filtering, grouping, and handling of missing data.

### Relationship Between Series and DataFrame
- A **Series** is a single column of data.
- A **DataFrame** is a collection of multiple Series joined together sharing the same index.

## Step 2 – Demonstrate the Concept

### Real-World Analogy
Think of an employee database inside a company.

The database stores employee details like `Employee ID`, `Name`, `Department`, and `Salary` in an organized table.

Each row represents an individual employee, and each column represents a specific detail about them.

### Simple Example
Suppose we have student details:

| Name | Mark | Grade |
|------|------|-------|
| Anbu | 85   | A     |
| Kavya| 92   | A+    |
| Suresh| 78  | B     |

Here, `Name`, `Mark`, and `Grade` are column names, and each row stores details for one student.

### Business Example
In a retail store management system, a DataFrame stores transaction logs:
- Columns: `Order_ID`, `Product_Name`, `Quantity`, `Price`, `Payment_Status`
- Rows: Individual purchase transactions made by customers.

Business analysts can filter this DataFrame to find unpaid orders or calculate total sales revenue.

### AI/ML Use Case
Data science models require structured input features. A DataFrame holds the complete training dataset where:
- Feature columns ($X$) represent attributes (like `Age`, `Income`, `Credit Score`).
- The Target column ($y$) represents what the model needs to predict (like `Loan Status`).

````python
# Import Pandas
import pandas as pd

# Create student data dictionary
student_data = {
    "Name": ["Anbu", "Kavya", "Suresh"],
    "Mark": [85, 92, 78],
    "Grade": ["A", "A+", "B"]
}

# Create a DataFrame
student_df = pd.DataFrame(student_data)

# Display the DataFrame
print("--- Full DataFrame ---")
print(student_df)

# Display a single column (Series) from the DataFrame
print("\n--- Mark Column Only ---")
print(student_df["Mark"])
````

## Step 3 – Explanation

### Code Explanation
- `import pandas as pd` imports the Pandas library into Python.
- `student_data` is a dictionary where keys become column names and lists become column values.
- `pd.DataFrame(student_data)` converts the dictionary into a tabular structure.
- `student_df` stores the created DataFrame object.
- `print(student_df)` prints the complete multi-column table.
- `student_df["Mark"]` selects and prints only the `"Mark"` column.

### What the Code Does
The code organizes multi-attribute data into a structured 2D table and extracts a specific column from it.

### Why `pd.DataFrame()` Is Used
`pd.DataFrame()` is used to construct 2-dimensional tabular structures that manage datasets containing multiple columns and mixed data types.

### Expected Output
```text
--- Full DataFrame ---

1   Kavya    92    A+
2  Suresh    78     B

--- Mark Column Only ---
0    85

2    78
Name: Mark, dtype: int64

### Where It Is Used in AI/ML or Data Analytics
Dataset Ingestion: In Machine Learning and Analytics, raw datasets (CSV, Excel, SQL) are loaded into DataFrames as the fundamental starting object.

Exploratory Data Analysis (EDA): DataFrames are used to summarize statistics, inspect column values, and check data distributions before modeling.

# 5. Creating DataFrames

## Step 1 – Understand the Concept

Creating DataFrames is the fundamental first step when working with Pandas in Python.

Data can originate from various sources like Python data structures, files, or database queries. Pandas provides multiple ways to build a DataFrame.

### Common Ways to Create a DataFrame
1. **From a Dictionary of Lists** → Keys become column names, and list elements become row values (most common method).
2. **From a List of Dictionaries** → Each dictionary represents a single row, where keys are column names.
3. **From a List of Lists** → Raw 2D nested lists converted to a table by explicitly providing column names.
4. **From a NumPy Array** → Numerical grids converted to a DataFrame with custom column headers.

### Key Rules When Creating DataFrames
- All lists inside a dictionary must have the **same length** (equal number of rows).
- If column names are not specified for list-based inputs, Pandas automatically assigns integer headers (`0, 1, 2...`).

## Step 2 – Demonstrate the Concept

### Real-World Analogy
Think of assembling a customer directory:
- **Dictionary of Lists:** You collect all customer names in one list, all phone numbers in another list, and combine them into a single ledger.
- **List of Dictionaries:** You fill out individual customer record index cards (one per customer) and stack them into a single binder.

Both methods result in the exact same ledger book (DataFrame).

### Simple Example
Suppose we want to record product information:

| Product | Price | Stock |
|---------|-------|-------|
| Laptop  | 55000 | 10    |
| Mouse   | 500   | 50    |
| Keyboard| 1200  | 30    |

We can create this table in Python using a Dictionary of Lists or a List of Dictionaries.

### Business Example
An online retail platform receives real-time API responses in JSON format (List of Dictionaries). 

Pandas converts these API records directly into a DataFrame for order processing and inventory updates.

### AI/ML Use Case
When creating synthetic datasets or loading preprocessed arrays from NumPy models, data scientists construct custom DataFrames to label features before feeding data into ML pipelines.

````python
# Import Pandas
import pandas as pd

# Method 1: Creating a DataFrame from a Dictionary of Lists
data_dict = {
    "Product": ["Laptop", "Mouse", "Keyboard"],
    "Price": [55000, 500, 1200],
    "Stock": [10, 50, 30]
}
df_from_dict = pd.DataFrame(data_dict)

print("--- 1. DataFrame from Dictionary of Lists ---")
print(df_from_dict)


# Method 2: Creating a DataFrame from a List of Dictionaries
data_list_of_dicts = [
    {"Product": "Laptop", "Price": 55000, "Stock": 10},
    {"Product": "Mouse", "Price": 500, "Stock": 50},
    {"Product": "Keyboard", "Price": 1200, "Stock": 30}
]
df_from_list = pd.DataFrame(data_list_of_dicts)

print("\n--- 2. DataFrame from List of Dictionaries ---")
print(df_from_list)
````

## Step 3 – Explanation

### Code Explanation
- `import pandas as pd` imports the Pandas library.
- `data_dict` creates column-oriented data where each key is a column name.
- `pd.DataFrame(data_dict)` builds the table using column lists.
- `data_list_of_dicts` creates row-oriented data where each dictionary represents one record.
- `pd.DataFrame(data_list_of_dicts)` builds the table row by row, automatically aligning key names into columns.

### What the Code Does
The code demonstrates two different ways to convert standard Python data structures into a clean 2D Pandas DataFrame table.

### Why `pd.DataFrame()` Is Used
`pd.DataFrame()` acts as a flexible constructor that accepts various Python formats (dictionaries, lists, tuples, NumPy arrays) and transforms them into a structured tabular format.

### Expected Output
```text
--- 1. DataFrame from Dictionary of Lists ---
    Product  Price  Stock
0    Laptop  55000     10
1     Mouse    500     50
2  Keyboard   1200     30

--- 2. DataFrame from List of Dictionaries ---
    Product  Price  Stock
0    Laptop  55000     10
1     Mouse    500     50
2  Keyboard   1200     30

### Where It Is Used in AI/ML or Data Analytics
API Data Parsing: Web APIs often return data in JSON (List of Dictionaries format). pd.DataFrame() instantly parses it into an analytical table.

Custom Feature Table Creation: Data engineers build manual DataFrames during mock dataset creation, baseline tests, or custom feature generator outputs.

Data Transformation Pipelines: Converting arrays produced by preprocessing tools back into labeled DataFrames for easy interpretation.

# 6. Data Types 

## Step 1 – Understand the Concept

Data Types (or `dtypes` in Pandas) define what kind of value a column holds, such as numbers, text, or true/false values.

Understanding data types is essential because mathematical operations and functions depend on the correct column type.

### Common Data Types in Pandas
1. **`int64`** → Whole numbers without decimals (e.g., `25`, `100`, `1500`).
2. **`float64`** → Numbers with decimal points (e.g., `25.5`, `99.99`, `1800.0`).
3. **`object`** → Text/String values or mixed data types (e.g., `"Arun"`, `"Laptop"`).
4. **`bool`** → True or False logical values (`True`, `False`).
5. **`datetime64`** → Dates and timestamps (e.g., `2026-10-06`).

### Checking Data Types
In Pandas, we use the `.dtypes` attribute on a DataFrame to inspect the data type of every column at once.

## Step 2 – Demonstrate the Concept

### Real-World Analogy
Think of a registration form:
- **Age field** expects numbers only (`int64`).
- **Price field** expects decimals (`float64`).
- **Name field** expects letters and text (`object`).
- **Terms Accepted field** expects a checkbox tick (`bool`).

If someone writes text in the age box, the system gives an error. Similarly, Pandas keeps track of types for each column to ensure calculations are accurate.

### Simple Example
Suppose we have employee records:

| Name (`object`) | Age (`int64`) | Salary (`float64`) | Is_FullTime (`bool`) |
|-----------------|---------------|--------------------|----------------------|
| Arun            | 28            | 55000.50           | True                 |
| Priya           | 32            | 72000.00           | False                |

Pandas automatically assigns the appropriate data type to each column based on its values.

### Business Example
In a financial application, transaction amounts stored accidentally as text (`object` type) will fail when calculating total revenue. Identifying and casting column types ensures correct financial summaries.

### AI/ML Use Case
Machine learning models cannot process text (`object`) directly. Data scientists check `.dtypes` to identify string/categorical columns so they can convert them into numbers using techniques like One-Hot Encoding before model training.

````python
# Import Pandas
import pandas as pd

# Create dataset with various data types
employee_data = {
    "Name": ["Arun", "Priya", "Ravi"],
    "Age": [28, 32, 24],
    "Salary": [55000.50, 72000.00, 48000.75],
    "Is_FullTime": [True, False, True]
}

# Create DataFrame
employee_df = pd.DataFrame(employee_data)

# Display the DataFrame
print("--- Employee DataFrame ---")
print(employee_df)

# Check data types of all columns
print("\n--- Data Types of Columns ---")
print(employee_df.dtypes)
````

## Step 3 – Explanation

### Code Explanation
- `import pandas as pd` imports the Pandas library into Python.
- `employee_data` contains text, integer, float, and boolean values in separate key lists.
- `employee_df = pd.DataFrame(employee_data)` creates the DataFrame table.
- `employee_df.dtypes` accesses the column metadata and returns the assigned data type for each column.

### What the Code Does
The code builds a table containing multiple data types and inspects the internal Pandas data type assigned to each column.

### Why `.dtypes` Is Used
`.dtypes` is used to quickly check column data types so data scientists can verify if data is ready for numerical operations or requires type conversion.

### Expected Output
```text
--- Employee DataFrame ---
    Name  Age    Salary  Is_FullTime
0   Arun   28  55000.50         True
1  Priya   32  72000.00        False
2   Ravi   24  48000.75         True

--- Data Types of Columns ---
Name            object
Age              int64
Salary         float64
Is_FullTime       bool
dtype: object

### Where It Is Used in AI/ML or Data Analytics
Data Type Validation: Before performing math or statistics, analysts run .dtypes to make sure numeric columns aren't accidentally saved as strings.

Preprocessing & Feature Engineering: Identifies text/categorical columns that need to be transformed into numerical formats before feeding data to AI/ML algorithms.

Memory Optimization: Converting large int64 or float64 columns to smaller types (like int32 or float32) helps reduce RAM usage when working with big datasets.

# 7 – Index

## Step 1 – Understand the Concept

An index is a label used to identify each row in a Pandas Series or DataFrame.

By default, Pandas assigns numerical indexes such as 0, 1, 2, 3, and so on.

We can also create custom indexes using meaningful labels such as Customer IDs, Employee IDs, or Product IDs.

The index helps us identify and access specific rows easily.

Commonly used:
- `index=` – used to assign custom index labels while creating a DataFrame.
- `.index` – used to view the index labels of a DataFrame or Series.

## Step 2 – Demonstrate the Concept

### Real-World Analogy

Think of a college class where every student has a unique roll number. The roll number helps us identify each student easily. In the same way, an index identifies each row in a DataFrame.

### Simple Example

A DataFrame may have default indexes such as 0, 1, and 2. We can replace them with meaningful labels such as Student1, Student2, and Student3.

### Business Example

A company can use Employee IDs such as E001, E002, and E003 as indexes. This makes it easier to identify employee records.

### AI/ML Use Case

In AI/ML and Data Analytics, indexes help identify and retrieve individual observations or records from a dataset. For example, customer IDs or transaction IDs can be used to locate specific records during data analysis.

````python
# Import Pandas
import pandas as pd

# Create employee data
employee_data = {
    "Name": ["Arun", "Priya", "Ravi"],
    "Age": [25, 28, 24]
}

# Create a DataFrame with custom indexes
employee_df = pd.DataFrame(
    employee_data,
    index=["E001", "E002", "E003"]
)

# Display the DataFrame
print(employee_df)
````

## Step 3 – Code Explanation

### Code Explanation

First, we import Pandas using `import pandas as pd`.

Then, we create a dictionary containing employee names and ages.

Next, we create a DataFrame using `pd.DataFrame()`.

The `index=` parameter is used to assign custom labels:
- E001 → Arun
- E002 → Priya
- E003 → Ravi

Finally, `print()` displays the DataFrame with the custom index.

### What the Code Does

The code creates a DataFrame where each employee row has a meaningful Employee ID instead of the default numerical index.

### Why the `index=` Parameter Is Used

The `index=` parameter allows us to assign custom labels to DataFrame rows.

This is useful when the data already has meaningful identifiers such as Employee ID, Customer ID, or Product ID.

### Expected Output

```text
     Name  Age
E001 Arun   25
E002 Priya  28
E003 Ravi   24

Here, E001, E002, and E003 are the index labels used to identify each employee record.

### Where It Is Used in AI/ML or Data Analytics

Indexes are useful for identifying and retrieving specific observations in datasets. In Data Analytics, customer IDs or transaction IDs can help locate records during analysis. In AI/ML projects, indexes can also help track individual data records before or after data preprocessing.

#9 – Reading Data

## Step 1 – Understand the Concept

Reading Data means loading data from an external file into a Pandas DataFrame.

Pandas provides different functions to read different file formats.

Commonly used functions:
- `pd.read_csv()` – reads data from a CSV file.
- `pd.read_excel()` – reads data from an Excel file.
- `pd.read_json()` – reads data from a JSON file.

After reading the data, we can use Pandas functions and methods to clean, analyze, and process the DataFrame.

## Step 2 – Demonstrate the Concept

### Real-World Analogy

````python
# Import Pandas and StringIO (StringIO lets us simulate a CSV file inside code)
import pandas as pd
from io import StringIO

# 1. Simulating a CSV file in memory (acts just like a real file saved on your computer)
simulated_csv_file = """Name,Age,City,Salary
Arun,25,Chennai,45000
Priya,28,Bangalore,55000
Ravi,24,Hyderabad,40000
Kavya,30,Chennai,60000
Deepa,27,Bangalore,50000"""

# 2. Reading the CSV data into a Pandas DataFrame
# (In real life, you will write: df = pd.read_csv('your_file.csv'))
df = pd.read_csv(StringIO(simulated_csv_file))

# 3. Viewing the loaded data
print("--- Loaded DataFrame from CSV ---")
print(df)

# 4. Checking the first 2 rows
print("\n--- First 2 Rows (df.head(2)) ---")
print(df.head(2))

# 5. Checking table size (Rows, Columns)
print("\nDataset Shape (Rows, Columns):", df.shape)
````

- `pd.read_csv(...)` reads the comma-separated data and converts it into a neat 2D DataFrame table.
- `df.head(2)` shows only the top 2 rows from the loaded dataset.
- `df.shape` gives us a tuple showing `(total rows, total columns)`.

The code loads external CSV-formatted data into Python, turns it into a DataFrame, and inspects its top rows and size.

#10 – Writing Data

## Step 1 – Understand the Concept

Writing Data means saving a Pandas DataFrame into an external file.

After cleaning, modifying, or analyzing data, we may need to save the updated DataFrame for future use.

Commonly used functions:
- `df.to_csv()` – saves a DataFrame as a CSV file.
- `df.to_excel()` – saves a DataFrame as an Excel file.
- `df.to_json()` – saves a DataFrame as a JSON file.

The `index=False` parameter can be used when we do not want to save the DataFrame index as a separate column.

## Step 2 – Demonstrate the Concept

### Real-World Analogy

Imagine writing information in a notebook and then saving that notebook for future use. Similarly, writing data saves a DataFrame into a file so that we can use it later.

### Simple Example

Suppose we create a DataFrame containing product names and prices. We can use `df.to_csv()` to save this data into a CSV file.

### Business Example

A company may analyze daily sales data and calculate the final sales amount. The updated sales DataFrame can be saved as a CSV or Excel file and shared with other teams.

### AI/ML Use Case

In AI/ML projects, processed datasets and prediction results often need to be saved for later use. For example, a cleaned dataset or model prediction results can be exported to a CSV file for further analysis or reporting.

````python
# Import Pandas
import pandas as pd

# Create product data
product_data = {
    "Product": ["Laptop", "Phone", "Tablet"],
    "Price": [50000, 20000, 30000]
}

# Create a DataFrame
product_df = pd.DataFrame(product_data)

# Write the DataFrame to a CSV file
product_df.to_csv("products.csv", index=False)

print("Data saved successfully.")
````

## Step 3 – Code Explanation

### Code Explanation

First, we import Pandas using `import pandas as pd`.

Then, we create product data using a Python dictionary.

Next, we create a DataFrame using `pd.DataFrame()`.

The `to_csv()` method saves the DataFrame into a CSV file named `products.csv`.

The `index=False` parameter prevents the DataFrame index from being saved as an additional column.

Finally, `print()` displays a confirmation message.

### What the Code Does

The code creates a product DataFrame and saves it as a CSV file named `products.csv`.

### Why the `to_csv()` Method Is Used

The `to_csv()` method is used to export a Pandas DataFrame into CSV format.

The `index=False` parameter is used because the default Pandas index is not required as a separate column in the saved file.
