# String Operations

## Step 1: Understand the Concept

The `lower()` method converts all uppercase letters in a string into lowercase letters.

In Pandas, `.str.lower()` is used to convert all string values in a Series into lowercase.

### Why Is It Used?

- To standardize text data.
- To make text comparisons consistent.
- To clean customer names, email addresses, and product categories.
- To reduce mismatches caused by different letter cases.

### Syntax

For a Python string:

text.lower()

For a Pandas Series:

dataframe["Column_Name"].str.lower()

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine writing the same person's name in different ways:

- PRIYA
- Priya
- priya

Using `lower()`, all three can be converted to `priya`.

### Simple Example

| Original Text | After lower() |
|---|---|
| HELLO | hello |
| Python | python |
| DATA ANALYTICS | data analytics |

### Business Example

An online shopping company receives product categories such as ELECTRONICS, Electronics, and electronics.

Converting these values to lowercase helps standardize the category text before analysis.

### AI/ML Use Case

In text preprocessing, lowercase conversion helps standardize words before tasks such as text classification and sentiment analysis.

For example, "GOOD" and "good" can be converted into the same form to reduce case-based differences.

```python
import pandas as pd

# Create a sample DataFrame
customer_df = pd.DataFrame({
    "Customer_Name": ["PRIYA", "Haripriya", "KUMAR", "ANU"]
})

# Convert customer names to lowercase
customer_df["Customer_Name_Lower"] = (
    customer_df["Customer_Name"].str.lower()
)

print(customer_df)
```

## Step 3: Code Explanation

### Code Explanation

First, we import Pandas and create a DataFrame containing customer names.

Then, `.str.lower()` converts every string value in the Customer_Name column into lowercase.

The converted values are stored in a new column named Customer_Name_Lower.

### What the Code Does

The code:
1. Creates a DataFrame with customer names.
2. Converts all names to lowercase.
3. Stores the converted names in a new column.
4. Displays the updated DataFrame.

### Why the Function and Attribute Are Used

- `pd.DataFrame()` creates a tabular dataset.
- `.str` provides string operations for values in a Pandas Series.
- `.lower()` converts uppercase letters into lowercase letters.

Creating a new column preserves the original names for comparison.

### Expected Output

| Customer_Name | Customer_Name_Lower |
|---|---|
| PRIYA | priya |
| Haripriya | haripriya |
| KUMAR | kumar |
| ANU | anu |

The new column contains lowercase versions of the original names.

### Where It Is Used in AI/ML or Data Analytics

- Text preprocessing
- Customer data cleaning
- Email and text standardization
- Sentiment analysis
- Text classification
- Natural Language Processing (NLP)

# upper()

- To prepare categorical text data for analysis.

text.upper()

dataframe["Column_Name"].str.upper()

# strip()

## Step 1: Understand the Concept

The `strip()` method removes unwanted spaces or specified characters from the beginning and end of a string.

In Pandas, `.str.strip()` is used to remove leading and trailing spaces from string values.

### Why Is It Used?

- To remove unwanted spaces.
- To clean text data.
- To standardize values before comparison.
- To prevent duplicate values caused by extra spaces.

### Syntax

For a Python string:

text.strip()

For a Pandas Series:

dataframe["Column_Name"].str.strip()

The `strip()` method does not remove spaces between words.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine writing a customer's name on a form with extra spaces:

"  Priya  "

The actual name is Priya, so the unnecessary spaces at the beginning and end should be removed.

`strip()` performs this cleaning.

### Simple Example

| Original Text | After strip() |
|---|---|
| "  Priya  " | "Priya" |
| "  Chennai" | "Chennai" |
| "Madurai  " | "Madurai" |

Spaces between words are not removed.

For example:

"New Delhi"

remains:

"New Delhi"

### Business Example

Customer data collected from forms or files may contain unwanted spaces.

For example:

" Chennai "

and

"Chennai"

may be treated as different values during analysis.

Using `strip()` helps standardize these values.

### AI/ML Use Case

In data preprocessing, unnecessary spaces can create inconsistent categorical values.

Removing leading and trailing spaces helps clean the data before analysis or machine learning.

```python
import pandas as pd

# Create a sample DataFrame with extra spaces
customer_df = pd.DataFrame({
    "Customer_Name": ["  Priya  ", " Haripriya", "Kumar  ", "  Anu"]
})

# Remove leading and trailing spaces
customer_df["Clean_Name"] = customer_df["Customer_Name"].str.strip()

print(customer_df)
```

## Step 3: Code Explanation

### Code Explanation

First, we create a DataFrame containing customer names with extra spaces.

Then, `.str.strip()` removes spaces from the beginning and end of each name.

The cleaned values are stored in a new column named Clean_Name.

### What the Code Does

The code:
1. Creates customer names with unwanted spaces.
2. Applies the `strip()` operation to each value.
3. Removes leading and trailing spaces.
4. Stores the cleaned names in a new column.
5. Displays the DataFrame.

### Why the Function and Attribute Are Used

- `pd.DataFrame()` creates the dataset.
- `.str` provides string operations for a Pandas Series.
- `.strip()` removes unwanted spaces from the beginning and end of strings.

A new column is used to preserve the original values.

### Expected Output

| Customer_Name | Clean_Name |
|---|---|
| "  Priya  " | "Priya" |
| " Haripriya" | "Haripriya" |
| "Kumar  " | "Kumar" |
| "  Anu" | "Anu" |

The Clean_Name column contains values without leading or trailing spaces.

# split()

## Step 1: Understand the Concept

The `split()` method divides a string into smaller parts based on a separator.

By default, `split()` separates a string using whitespace.

Example:

"Haripriya P"

becomes:

["Haripriya", "P"]

A specific separator can also be provided.

Example:

"Python,SQL,Pandas"

Using a comma as the separator produces:

["Python", "SQL", "Pandas"]

### Why Is It Used?

- To separate combined text values.
- To extract individual words or values.
- To process structured text.
- To prepare text for further analysis.

### Syntax

For a Python string:

text.split(separator)

For a Pandas Series:

dataframe["Column_Name"].str.split(separator)

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a person's full name:

"Haripriya P"

We can split it into smaller parts:

- Haripriya
- P

Similarly, a list such as:

"Python,SQL,Pandas"

can be split into individual skills.

### Simple Example

| Original Text | Separator | Result |
|---|---|---|
| "Haripriya P" | Space | ["Haripriya", "P"] |
| "Python,SQL,Pandas" | Comma | ["Python", "SQL", "Pandas"] |

### Business Example

An e-commerce dataset may contain product categories in a combined field such as:

"Electronics,Mobile,Accessories"

Using `split()` allows the company to separate these categories for further processing.

### AI/ML Use Case

In text preprocessing, `split()` can be used to break sentences or text into individual words.

For example:

"Machine Learning is useful"

can be split into individual words before further text processing.

```python
import pandas as pd

# Create a sample DataFrame
customer_df = pd.DataFrame({
    "Skills": [
        "Python,SQL,Pandas",
        "Python,NumPy,Matplotlib",
        "SQL,Pandas"
    ]
})

# Split the skills using a comma
customer_df["Skill_List"] = customer_df["Skills"].str.split(",")

print(customer_df)
```

## Step 3: Code Explanation

### Code Explanation

First, we create a DataFrame containing multiple skills in a single string.

Then, `.str.split(",")` separates each string wherever a comma appears.

The result is stored in a new column named Skill_List.

### What the Code Does

The code:
1. Creates a DataFrame containing combined skill values.
2. Uses a comma as the separator.
3. Splits each string into individual values.
4. Stores the resulting lists in a new column.
5. Displays the DataFrame.

### Why the Function and Attribute Are Used

- `pd.DataFrame()` creates the dataset.
- `.str` provides string operations for a Pandas Series.
- `.split(",")` separates the string wherever a comma is found.

The comma is used as the separator because the skills are stored using commas.

### Expected Output

The Skill_List column will contain:

["Python", "SQL", "Pandas"]

["Python", "NumPy", "Matplotlib"]

["SQL", "Pandas"]

Each original string has been converted into a list of individual skills.

### Where It Is Used in AI/ML or Data Analytics

- Text preprocessing
- Data cleaning
- Keyword extraction
- Customer preference analysis
- Skill analysis
- Natural Language Processing (NLP)

# contains()

## Step 1: Understand the Concept

The `contains()` method checks whether a string contains a specific text or pattern.

In Pandas, `.str.contains()` returns `True` when the specified text is found and `False` when it is not found.

### Why Is It Used?

- To search for specific text.
- To filter rows based on text.
- To find keywords in text data.
- To identify records containing a particular word or phrase.

### Syntax

For a Python string:

text.find("value")

For a Pandas Series:

dataframe["Column_Name"].str.contains("value")

The Pandas `str.contains()` method is commonly used for text searching and filtering.

    

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine searching through a list of customer comments to find comments that contain the word "good".

Instead of checking every comment manually, `contains()` can identify the matching comments.

### Simple Example

| Text | Contains "Python"? |
|---|---|
| Python Developer | True |
| Data Analyst | False |
| Python Programmer | True |

### Business Example

An e-commerce company collects customer reviews.

The company can search for reviews containing words such as:
- good
- bad
- delivery
- quality

This can help identify specific customer feedback.

### AI/ML Use Case

In text preprocessing, `contains()` can be used to identify records containing specific keywords.

For example, customer reviews containing "excellent" or "poor" can be selected for further sentiment analysis.

```python
import pandas as pd

# Create a sample DataFrame
customer_df = pd.DataFrame({
    "Review": [
        "Good product and fast delivery",
        "Average product quality",
        "Excellent product",
        "Good quality and service"
    ]
})

# Check which reviews contain the word "good"
customer_df["Contains_Good"] = customer_df["Review"].str.contains(
    "good",
    case=False
)

print(customer_df)
```

## Step 3: Code Explanation

### Code Explanation

First, we create a DataFrame containing customer reviews.

Then, `.str.contains()` checks whether each review contains the word "good".

The result is stored in a new column named Contains_Good.

### What the Code Does

The code:
1. Creates customer review data.
2. Searches for the word "good" in each review.
3. Returns `True` when the word is found.
4. Returns `False` when the word is not found.
5. Stores the results in a new column.

### Why the Function and Parameter Are Used

- `pd.DataFrame()` creates the dataset.
- `.str` provides string operations for a Pandas Series.
- `.contains()` searches for specific text.
- `case=False` makes the search case-insensitive.

Using `case=False` means both "Good" and "good" will be treated as a match.

### Expected Output

| Review | Contains_Good |
|---|---|
| Good product and fast delivery | True |
| Average product quality | False |
| Excellent product | False |
| Good quality and service | True |

The result is `True` for reviews containing the word "good", regardless of whether it starts with an uppercase or lowercase letter.

### Where It Is Used in AI/ML or Data Analytics

- Text filtering
- Customer review analysis
- Keyword detection
- Sentiment analysis preprocessing
- Natural Language Processing (NLP)
- Searching text datasets

# replace()

## Step 1: Understand the Concept

The `replace()` method replaces one part of a string with another value.

In Pandas, `.str.replace()` is used to replace text within string values in a Series.

### Why Is It Used?

- To clean text data.
- To correct unwanted text.
- To standardize values.
- To remove or replace specific characters.
- To correct inconsistent text values.

### Syntax

For a Python string:

text.replace("old_value", "new_value")

For a Pandas Series:

dataframe["Column_Name"].str.replace("old_value", "new_value")

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a customer address contains:

"Chennai City"

If the company wants to store only "Chennai", we can replace:

" City"

with an empty string.

The result becomes:

"Chennai"

### Simple Example

| Original Text | Replacement | Result |
|---|---|---|
| Chennai City | City → "" | Chennai |
| Python Developer | Developer → Programmer | Python Programmer |
| Data-Science | - → " " | Data Science |

### Business Example

A company may receive product names with inconsistent labels.

For example:

- Mobile-Phone
- Mobile Phone
- Mobile_Phone

Using `replace()`, unwanted characters or text can be standardized.

### AI/ML Use Case

In text preprocessing, `replace()` can be used to remove unwanted characters, correct text, or standardize values before further analysis.

For example, unwanted symbols can be removed before text is passed to an NLP pipeline.

```python
import pandas as pd

# Create a sample DataFrame
product_df = pd.DataFrame({
    "Product_Name": [
        "Mobile-Phone",
        "Laptop-Computer",
        "Wireless-Headphones"
    ]
})

# Replace hyphens with spaces
product_df["Clean_Product_Name"] = (
    product_df["Product_Name"].str.replace("-", " ", regex=False)
)

print(product_df)
```

## Step 3: Code Explanation

### Code Explanation

First, we create a DataFrame containing product names with hyphens.

Then, `.str.replace()` replaces each hyphen with a space.

The cleaned product names are stored in a new column named Clean_Product_Name.

### What the Code Does

The code:
1. Creates product names containing hyphens.
2. Searches for the hyphen character.
3. Replaces each hyphen with a space.
4. Stores the cleaned values in a new column.
5. Displays the DataFrame.

### Why the Function and Parameters Are Used

- `pd.DataFrame()` creates the dataset.
- `.str` provides string operations for a Pandas Series.
- `.replace("-", " ")` changes each hyphen into a space.
- `regex=False` tells Pandas to treat the hyphen as a normal text character rather than a regular expression pattern.

### Expected Output

| Product_Name | Clean_Product_Name |
|---|---|
| Mobile-Phone | Mobile Phone |
| Laptop-Computer | Laptop Computer |
| Wireless-Headphones | Wireless Headphones |

The hyphens have been replaced with spaces, making the product names cleaner and easier to read.

### Where It Is Used in AI/ML or Data Analytics

# Regular Expressions (Regex)

## Step 1: Understand the Concept

Regular Expression (Regex) is a pattern used to search, match, extract, or validate text.

Regex helps us identify whether text follows a specific pattern.

For example, a 10-digit phone number can be represented using the pattern:

^\d{10}$

Important Regex symbols:

- `^` → Matches the start of the text.
- `\d` → Matches any digit from 0 to 9.
- `{10}` → Matches exactly 10 occurrences.
- `$` → Matches the end of the text.

Why do we use Regex?

- To find specific patterns in text.
- To validate data formats.
- To extract useful information from text.
- To clean text during data preprocessing.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine checking a list of phone numbers manually to find which numbers have exactly 10 digits.

Instead of checking each number manually, Regex allows us to define the required pattern and check all values automatically.

### Simple Example

Suppose we have these phone numbers:

| Phone Number | Result |
|---|---|
| 9876543210 | Valid |
| 12345 | Invalid |
| 8765432109 | Valid |
| 98765 | Invalid |

The pattern `^\d{10}$` checks whether the complete value contains exactly 10 digits.

### Business Example

A company may store thousands of customer phone numbers.

Regex can be used to identify phone numbers that do not follow the required format before storing or processing the data.

### AI/ML Use Case

Regex is useful in text preprocessing for:

- Extracting phone numbers, email addresses, and IDs.
- Removing unwanted characters.
- Finding specific patterns in text data.
- Preparing text data before NLP processing.

```python
import pandas as pd

customer_df = pd.DataFrame({
    "Phone": [
        "9876543210",
        "12345",
        "8765432109",
        "98765"
    ]
})

customer_df["Valid_Phone"] = customer_df["Phone"].str.contains(
    r"^\d{10}$",
    regex=True
)

print(customer_df)
```

## Step 3: Code Explanation

### Code Explanation

`pd.DataFrame()` creates a table containing customer phone numbers.

`.str.contains()` checks whether each phone number matches the given Regex pattern.

The Regex pattern `r"^\d{10}$"` means:

- `^` → Start of the text
- `\d` → Any digit from 0 to 9
- `{10}` → Exactly 10 digits
- `$` → End of the text

`regex=True` tells Pandas to treat the given pattern as a Regular Expression.

### What the Code Does

The code checks every phone number and creates a new column called `Valid_Phone`.

It returns `True` when the phone number contains exactly 10 digits and `False` otherwise.

### Why the Function/Method Is Used

- `pd.DataFrame()` → Creates the dataset.
- `.str.contains()` → Searches for a pattern in string values.
- `regex=True` → Enables Regular Expression pattern matching.

### Expected Output

The output will look like:

       Phone  Valid_Phone
0  9876543210         True
1       12345        False
2  8765432109         True
3       98765        False

`True` means the value follows the required 10-digit pattern.

`False` means the value does not follow the required pattern.

### Where It Is Used in AI/ML or Data Analytics

Regex is used in data cleaning and text preprocessing to validate, extract, and clean structured patterns from text data before analysis or model training.
