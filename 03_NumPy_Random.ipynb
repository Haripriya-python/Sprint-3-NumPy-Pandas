# 1. Random Numbers

## Step 1 – Understand the Concept

Random numbers are values generated in a random or pseudo-random manner.

NumPy provides the `np.random` module for generating random numbers.

The `np.random.rand()` function generates random values between 0 and 1.

### Syntax

`np.random.rand()`

### Example

`np.random.rand()` may generate a value such as `0.6382`.

The generated value can be different each time the code is executed.

## Step 2 – Demonstrate the Concept

### Real-World Analogy

Rolling a dice gives a value randomly from a set of possible values.

### Simple Example

A random number generated by `np.random.rand()` can be `0.25`, `0.67`, or `0.91`.

### Business Example

A business can use random numbers to simulate customer behavior or generate sample data for testing.

### AI/ML Use Case

Random numbers are used in AI/ML for creating sample data, initializing model parameters, and performing randomized experiments.

````python
# Import NumPy
import numpy as np

# Generate a random number between 0 and 1
random_number = np.random.rand()

# Display the random number
print(random_number)
````

## Step 3 – Explanation

### Code Explanation

- `import numpy as np` imports the NumPy library.
- `np.random.rand()` generates a random number between 0 and 1.
- `random_number` stores the generated value.
- `print(random_number)` displays the random number.

### What the Code Does

The code generates one random number between 0 and 1.

### Why `np.random.rand()` Is Used

`np.random.rand()` is used when we need random floating-point values between 0 and 1.

### Expected Output

The output will be a random decimal value between 0 and 1.

For example:

`0.63824517`

The exact value can change each time the code runs.

### Where It Is Used in AI/ML or Data Analytics

Random numbers are used for generating sample data, simulations, random initialization, experiments, and other randomized processes in AI/ML and Data Analytics.

# 2. Random Seed

## Step 1 – Understand the Concept

A random seed is a starting value used by a random number generator.

Normally, random functions can generate different values each time the program runs.

By setting the same random seed, we can generate the same sequence of random values every time.

NumPy provides the `np.random.seed()` function to set the random seed.

### Syntax

`np.random.seed(seed_value)`

### Example

`np.random.seed(42)`

Here, `42` is the seed value.

The seed value can be any suitable integer. The important point is that using the same seed produces the same sequence of random values.

### Why Random Seed Is Important

Random seeds are useful when we need reproducible results. Reproducibility means getting the same result when the same code is executed again under the same conditions.

## Step 2 – Demonstrate the Concept

### Real-World Analogy

Imagine a game that randomly selects numbers. If we start the game with the same special starting key, the game can produce the same sequence of numbers.

The starting key is similar to a random seed.

### Simple Example

Without a seed:

`np.random.rand()`

The generated value can be different every time the program runs.

With a seed:

`np.random.seed(42)`

`np.random.rand()`

The same random value will be generated whenever the same seed is set before the random operation.

### Business Example

Suppose a company creates random sample customer data for testing. If the company needs to repeat the same analysis later, using a random seed allows the same sample data to be generated again.

### AI/ML Use Case

Random seeds are commonly used in AI/ML experiments. Many machine learning processes involve randomness, such as data splitting, model initialization, and sampling.

Setting a random seed helps developers and data scientists reproduce experiments and compare results consistently.
    

````python
# Import NumPy
import numpy as np

# Set the random seed
np.random.seed(42)

# Generate random numbers
random_numbers = np.random.rand(5)

# Display the random numbers
print(random_numbers)
````

## Step 3 – Explanation

### Code Explanation

- `import numpy as np` imports the NumPy library.
- `np.random.seed(42)` sets the random seed to 42.
- `np.random.rand(5)` generates five random floating-point numbers between 0 and 1.
- `random_numbers` stores the generated values.
- `print(random_numbers)` displays the random numbers.

### What the Code Does

The code first sets a fixed random seed and then generates five random numbers.

Because the seed is fixed at 42, the same sequence of five numbers will be generated whenever the same code is executed.

### Why `np.random.seed()` Is Used

`np.random.seed()` is used to make random results reproducible.

Without setting a seed, random values can change each time the program runs.

With the same seed, the same sequence of random values can be generated again.

### Why `42` Is Used

`42` is a commonly used example seed value. There is nothing special about the number 42.

Other integer values such as `10`, `100`, or `2026` can also be used as seed values.

### Expected Output

The output will be:

`[0.37454012 0.95071431 0.73199394 0.59865848 0.15601864]`

When the same seed and code are used, the same sequence will be produced.

### Where It Is Used in AI/ML or Data Analytics

Random seeds are important for reproducible AI/ML experiments. They can help produce consistent results when randomly splitting datasets, sampling data, initializing model parameters, or performing randomized experiments.

# 3. Random Integers

## Step 1 – Understand the Concept

Random integers are randomly generated whole numbers.

NumPy provides the `np.random.randint()` function to generate random integer values within a specified range.

### Syntax

`np.random.randint(low, high, size)`

### Parameters

- `low` → The starting value of the range.
- `high` → The upper limit of the range. This value is not included.
- `size` → The number or shape of random integers to generate.

### Example

`np.random.randint(1, 10)`

This generates one random integer from 1 to 9.

`np.random.randint(1, 10, 5)`

This generates five random integers from 1 to 9.

### Important Point

The `high` value is excluded from the generated results.

## Step 2 – Demonstrate the Concept

### Real-World Analogy

Rolling a dice is a simple example of generating a random integer. A dice can produce a whole number from 1 to 6.

In NumPy, this can be simulated using:

`np.random.randint(1, 7)`

The value 7 is excluded, so the possible results are 1 to 6.

### Simple Example

`np.random.randint(1, 10, 5)` can generate five random integers such as:

`[3, 8, 1, 6, 4]`

The exact values can change each time the code runs.

### Business Example

A company can generate random customer IDs, randomly select products for testing, or create sample integer-based business data.

For example, random quantities between 1 and 10 can be generated for testing an order-processing system.

### AI/ML Use Case

Random integers can be used to generate sample data, randomly select indexes, create test datasets, and perform randomized sampling during AI/ML experiments.

````python
# Import NumPy
import numpy as np

# Generate five random integers from 1 to 9
random_integers = np.random.randint(1, 10, 5)

# Display the random integers
print(random_integers)
````

## Step 3 – Explanation

### Code Explanation

- `import numpy as np` imports the NumPy library.
- `np.random.randint(1, 10, 5)` generates five random integers.
- `1` is the starting value.
- `10` is the upper limit and is not included.
- `5` specifies that five random integers should be generated.
- `random_integers` stores the generated values.
- `print(random_integers)` displays the result.

### What the Code Does

The code generates five random whole numbers between 1 and 9.

Because a seed of 42 is used, running the same code again produces the same sequence.

### Why `np.random.randint()` Is Used

`np.random.randint()` is used when we need random whole numbers within a specific range.

It is useful when decimal values are not required.

### Expected Output

`[7 4 8 5 7]`

These are five randomly generated integers between 1 and 9.

### Important Point

In `np.random.randint(1, 10, 5)`, the value `1` is included but `10` is excluded.

Therefore, the possible values are:

`1, 2, 3, 4, 5, 6, 7, 8, 9`

### Where It Is Used in AI/ML or Data Analytics

Random integers are used for random sampling, generating test data, selecting random indexes, simulations, and creating randomized datasets for AI/ML and Data Analytics.

# 4. Random Choice

## Step 1 – Understand the Concept

Random choice means selecting one or more elements randomly from an array or sequence.

NumPy provides the `np.random.choice()` function for random selection.

### Syntax

`np.random.choice(array, size=None, replace=True, p=None)`

- `array` → The source array from which values are selected.
- `size` → The number of values to select.
- `replace=True` → A selected value can be selected again.
- `replace=False` → A selected value cannot be selected again.
- `p` → Optional probability for each value.

### Example

If an array contains:

`["Apple", "Banana", "Mango"]`

`np.random.choice()` can randomly select one of these values.

For example, the result may be:

`"Banana"`

The result can be different in another execution because the selection is random.

## Step 2 – Demonstrate the Concept

### Real-World Analogy

Imagine writing four names on separate pieces of paper and randomly picking one paper from a box.

The selected name is unpredictable. In the same way, `np.random.choice()` randomly selects values from an array.

### Simple Example

Suppose we have three fruits:

`["Apple", "Banana", "Mango"]`

If we randomly select one fruit, the result can be Apple, Banana, or Mango.

If we select multiple values with `replace=True`, the same fruit can appear more than once.

### Business Example

An online shopping company may have thousands of customers.

If the company wants to randomly select one customer for a feedback survey, `np.random.choice()` can be used to select the customer randomly.

This helps avoid manually selecting customers and can provide a simple random sample.

### AI/ML Use Case

In AI/ML, random choice can be used to randomly select samples from a dataset.

For example, if a dataset contains many customer records, random samples can be selected for testing, analysis, or experimentation.

Random sampling can help create representative subsets of data and support model development and evaluation.

````python
# Import NumPy
import numpy as np

# Create an array of products
products = np.array(["Laptop", "Phone", "Tablet", "Headphones"])

# Select one random product
selected_product = np.random.choice(products)

# Display the selected product
print(selected_product)
````

## Step 3 – Explanation

### Code Explanation

- `import numpy as np` imports the NumPy library.
- `products = np.array(...)` creates a NumPy array containing four products.
- `np.random.choice(products)` randomly selects one product from the array.
- `print(selected_product)` displays the selected product.

### What the Code Does

The code creates a list of four products and randomly selects one product.

Because the seed is set to `42`, the same code produces the same result when executed again.

### Why `np.random.choice()` Is Used

`np.random.choice()` is used when we need to randomly select one or more values from an array.

It is useful for random sampling and selection.

### Why `np.random.seed(42)` Is Used

The seed makes the random result reproducible.

Using the same seed and the same code allows us to get the same random selection again.

### Expected Output

```text
Tablet

The selected product is Tablet for this example.

### Where It Is Used in AI/ML or Data Analytics

Random choice can be used to select random samples from datasets, create test samples, perform experiments, and support random sampling during AI/ML model development and data analysis.

# 5. Shuffle

## Step 1 – Understand the Concept

Shuffling means randomly changing the order of elements in an array.

NumPy provides the `np.random.shuffle()` function for shuffling arrays.

### Syntax

`np.random.shuffle(array)`

The function changes the order of the elements randomly.

### Important Point

`np.random.shuffle()` modifies the original array directly.

It does not return a new shuffled array.

### Example

Original array:

`[10, 20, 30, 40, 50]`

After shuffling, it may become:

`[30, 10, 50, 20, 40]`

The values remain the same, but their order changes.

## Step 2 – Demonstrate the Concept

### Real-World Analogy

Imagine five students standing in a line.

Before shuffling:

`Student 1 → Student 2 → Student 3 → Student 4 → Student 5`

If the teacher randomly changes their positions, the students may stand in a different order.

This is similar to how `np.random.shuffle()` changes the order of elements in an array.

### Simple Example

Suppose we have numbers:

`[1, 2, 3, 4, 5]`

After shuffling, the order may become:

`[4, 1, 5, 2, 3]`

The numbers are not changed or removed. Only their positions are changed.

### Business Example

An e-commerce company may have a list of customer records.

Before analysis or testing, the records can be shuffled so that they are not processed in their original order.

This can help reduce the effect of the original ordering of the data.

### AI/ML Use Case

In machine learning, data is often shuffled before training a model.

For example, if all positive examples are placed first and all negative examples are placed later, the model may learn from an unwanted ordering pattern.

Shuffling helps mix the data and provides a better random order before training or splitting the dataset.

````python
# Import NumPy
import numpy as np

# Create an array
numbers = np.array([10, 20, 30, 40, 50])

# Shuffle the array
np.random.shuffle(numbers)

# Display the shuffled array
print(numbers)
````

## Step 3 – Explanation

### Code Explanation

- `import numpy as np` imports the NumPy library.
- `numbers = np.array([10, 20, 30, 40, 50])` creates a NumPy array containing five numbers.
- `np.random.shuffle(numbers)` randomly changes the order of the elements in the array.
- `print(numbers)` displays the shuffled array.

### What the Code Does

The code creates an array containing five numbers and randomly changes their order.

The values remain the same, but their positions are changed.

For example, the original array is:

`[10, 20, 30, 40, 50]`

After shuffling, it could become:

`[30, 10, 50, 20, 40]`

The output can be different each time because the shuffle is random.

### Why `np.random.shuffle()` Is Used

`np.random.shuffle()` is used when we need to randomly change the order of elements in an array.

It is especially useful when the original order of data should not affect analysis or model training.

### Important Point

`np.random.shuffle()` modifies the original array directly.

For example:

`np.random.shuffle(numbers)`

changes `numbers` itself.

### Expected Output

The output will contain the same five numbers, but their order may be different.

Example:

`[30 10 50 20 40]`

### Where It Is Used in AI/ML or Data Analytics

Shuffling is commonly used before machine learning model training and dataset splitting.

It helps mix the observations so that the model does not depend on the original ordering of the data.

In Data Analytics, shuffling can also be useful when creating random samples or preparing data for experiments and testing.

# 6. Permutation

## Step 1 – Understand the Concept

Permutation means creating a new random arrangement of the elements in an array.

NumPy provides the `np.random.permutation()` function for this purpose.

### Syntax

`np.random.permutation(array)`

The function returns a new array with the elements arranged in a random order.

### Example

Original array:

`[10, 20, 30, 40, 50]`

A possible permutation is:

`[30, 10, 50, 20, 40]`

The original array remains unchanged.

### Shuffle vs Permutation

- `np.random.shuffle()` changes the original array.
- `np.random.permutation()` returns a new randomly arranged array.

This makes permutation useful when we want to keep the original data unchanged.

## Step 2 – Demonstrate the Concept

### Real-World Analogy

Imagine five students have five numbered positions.

If we randomly rearrange the students and write down their new positions without changing the original seating list, this is similar to permutation.

We get a new random arrangement while keeping the original arrangement available.

### Simple Example

Suppose we have:

`[1, 2, 3, 4, 5]`

A random permutation could be:

`[4, 2, 5, 1, 3]`

All five values are still present, but their positions have changed.

### Business Example

A company may have a list of customers and want to create a random ordering for a survey or experiment.

Using permutation allows the company to create a randomized order while keeping the original customer list unchanged.

### AI/ML Use Case

Permutation is useful when preparing data for machine learning experiments.

For example, a dataset can be randomly reordered before creating training and testing subsets.

Because permutation returns a new array, the original ordering can still be preserved for comparison or further processing.

````python
# Import NumPy
import numpy as np

# Create an array
numbers = np.array([10, 20, 30, 40, 50])

# Create a random permutation
permuted_numbers = np.random.permutation(numbers)

# Display the original array
print("Original array:", numbers)

# Display the permuted array
print("Permuted array:", permuted_numbers)
````

## Step 3 – Explanation

### Code Explanation

- `import numpy as np` imports the NumPy library.
- `numbers = np.array([10, 20, 30, 40, 50])` creates the original NumPy array.
- `np.random.permutation(numbers)` creates a new array with the elements arranged in a random order.
- `permuted_numbers` stores the new randomly arranged array.
- `print("Original array:", numbers)` displays the original array.
- `print("Permuted array:", permuted_numbers)` displays the new random arrangement.

### What the Code Does

The code creates an original array and then creates a randomly rearranged version of that array.

For example:

Original:

`[10 20 30 40 50]`

Permuted:

`[30 10 50 20 40]`

The values are the same, but their order is different.

### Why `np.random.permutation()` Is Used

It is used when we need a random ordering of data but want to keep the original array unchanged.

This is the main difference from `np.random.shuffle()`.

### Expected Output

The original array remains:

`[10 20 30 40 50]`

The permuted array will contain the same values in a random order.

Example:

`[30 10 50 20 40]`

The exact order may change each time.

### Where It Is Used in AI/ML or Data Analytics

Permutation can be used to randomly reorder datasets before machine learning training, testing, or experimentation.

It is also useful in Data Analytics when creating randomized samples or comparing results with different data orders while preserving the original dataset.

# 7. Random Distributions

## Step 1 – Understand the Concept

A random distribution describes how randomly generated values are spread according to a probability pattern.

NumPy provides different random distribution functions through the `np.random` module.

### Normal Distribution

One commonly used distribution is the Normal Distribution.

It is also called a Gaussian Distribution.

In a normal distribution, values are usually concentrated around the mean, while values farther from the mean occur less frequently.

### Function

`np.random.normal(loc, scale, size)`

- `loc` → Mean of the distribution.
- `scale` → Standard deviation of the distribution.
- `size` → Number of random values to generate.

### Example

`np.random.normal(50, 10, 5)`

This generates 5 random values from a normal distribution with:

- Mean = 50
- Standard deviation = 10
- Number of values = 5

The generated values will usually be around 50, but they can be lower or higher.

### Other Common Random Distributions

NumPy also provides distributions such as:

- Uniform Distribution
- Binomial Distribution
- Poisson Distribution
- Exponential Distribution

## Step 2 – Demonstrate the Concept

### Real-World Analogy

Imagine the heights of 1,000 people.

Most people may have heights around the average height, while very short and very tall people are less common.

When values are concentrated around the center and become less common toward the extremes, the data can resemble a normal distribution.

### Simple Example

Suppose the average exam mark is 70 and the standard deviation is 10.

We can generate random marks around this average using:

`np.random.normal(70, 10, 5)`

The generated values may look like:

`[65.2, 72.8, 81.1, 68.5, 75.4]`

The values are not exactly the same, but they are generally around the mean of 70.

### Business Example

A company may analyze customer spending.

If most customers spend around an average amount and fewer customers spend extremely low or extremely high amounts, the data may approximately follow a normal distribution.

Random values from a normal distribution can also be useful for simulations and testing.

### AI/ML Use Case

Normal distributions are commonly used in statistics and machine learning.

They can be used to simulate data, understand feature distributions, initialize model parameters, and perform statistical analysis.

Understanding distributions helps us understand how features are spread before building an AI/ML model.

````python
# Import NumPy
import numpy as np

# Generate random values from a normal distribution
random_values = np.random.normal(loc=50, scale=10, size=5)

# Display the random values
print(random_values)
````

## Step 3 – Explanation

### Code Explanation

- `import numpy as np` imports the NumPy library.
- `np.random.normal()` generates random values following a normal distribution.
- `loc=50` sets the mean of the distribution to 50.
- `scale=10` sets the standard deviation to 10.
- `size=5` tells NumPy to generate 5 random values.
- `random_values` stores the generated values.
- `print(random_values)` displays the generated values.

### What the Code Does

The code generates five random values from a normal distribution.

The center of the distribution is 50, so the generated values will generally be around 50.

Some values can be below 50 and some can be above 50.

The exact values change because they are randomly generated.

### Why `np.random.normal()` Is Used

`np.random.normal()` is used when we need to generate random data that follows a normal distribution.

The `loc` parameter controls the mean, while `scale` controls the standard deviation.

A larger `scale` creates values that are more spread out from the mean.

### Expected Output

An example output is:

`[54.96714153 48.61735699 56.47688538 65.23029856 47.65846625]`

The exact output can change each time.

### Where It Is Used in AI/ML or Data Analytics

Random distributions are used to simulate realistic data, understand feature distributions, perform statistical analysis, and support machine learning experiments.

Normal distributions are especially important in statistical modeling, data preprocessing, probability-based analysis, and some machine learning algorithms.
