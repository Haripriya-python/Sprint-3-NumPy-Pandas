# Performance Optimization - Vectorization

## Step 1: Understand the Concept

Vectorization is the process of performing an operation on an entire Series or array at once instead of processing each value individually using an explicit loop.

Vectorized operations are commonly provided by Pandas and NumPy.

Why do we use Vectorization?

- To perform operations efficiently.
- To reduce the need for explicit Python loops.
- To improve code readability.
- To improve performance when working with large datasets.

Example:

Instead of processing each price separately:

`for price in df["Price"]`

We can perform the operation directly:

`df["Price"] * 1.10`

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a shop has thousands of products and wants to increase every product price by 10%.

One approach is to take each product one by one and update its price.

Vectorization is like applying the same price update to the entire product list at once.

### Simple Example

Suppose the prices are:

| Product | Price |
|---|---:|
| Laptop | 50000 |
| Phone | 20000 |
| Tablet | 30000 |

To increase every price by 10%, we can use:

`Price * 1.10`

The operation is applied to the complete column.

### Business Example

An e-commerce company may have millions of product prices.

Instead of using a Python loop to update each price, a vectorized Pandas or NumPy operation can process the column efficiently.

### AI/ML Use Case

Vectorization is widely used in data preprocessing and feature engineering.

For example, numerical features can be scaled, transformed, or calculated using vectorized NumPy and Pandas operations before training a machine learning model.

````python
import pandas as pd

product_df = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Tablet"],
    "Price": [50000, 20000, 30000]
})

product_df["Updated_Price"] = product_df["Price"] * 1.10

print(product_df)
````

## Step 3: Code Explanation

### Code Explanation

`pd.DataFrame()` creates a product dataset with Product and Price columns.

The expression:

`product_df["Price"] * 1.10`

is a vectorized operation.

It multiplies every value in the Price column by 1.10 without explicitly writing a loop.

### What the Code Does

The code increases every product price by 10%.

For example:

- 50000 × 1.10 = 55000
- 20000 × 1.10 = 22000
- 30000 × 1.10 = 33000

The calculated values are stored in a new column called `Updated_Price`.

### Why the Function/Method Is Used

- `pd.DataFrame()` → Creates the dataset.
- `product_df["Price"]` → Selects the Price column.
- `* 1.10` → Increases every price by 10%.
- Vectorized operation → Applies the calculation to the complete column efficiently.

### Expected Output

The output will be similar to:

       Product  Price  Updated_Price
0       Laptop  50000        55000.0
1        Phone  20000        22000.0
2       Tablet  30000        33000.0

The `Updated_Price` column contains the prices after a 10% increase.

### Where It Is Used in AI/ML or Data Analytics

Vectorization is used for fast numerical calculations, data transformation, feature engineering, and preprocessing large datasets before analysis or machine learning.

# Performance Optimization - apply() vs Loops

## Step 1: Understand the Concept

Both loops and `apply()` can be used to perform operations on data.

A loop processes values one by one using Python iteration.

`apply()` allows us to apply a function to values in a Pandas Series or DataFrame.

Example using a loop:

```python
for price in df["Price"]:
    ...

df["Price"].apply(function)

### Why do we compare them?

To understand different ways of processing data.
To write cleaner Pandas code.
To choose an appropriate approach for a given operation.
To understand that vectorization is usually preferred for simple numerical operations.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a shop has a list of product prices and wants to label each product as "Expensive" or "Affordable".

Using a loop is like checking each product one by one.

Using `apply()` is like giving the same decision rule to Pandas and applying it to the column.

### Simple Example

Suppose:

| Product | Price |
|---|---:|
| Laptop | 50000 |
| Phone | 20000 |
| Mouse | 500 |

We can create a function that returns:

- "Expensive" if price is greater than 30000.
- "Affordable" otherwise.

This function can be applied to the Price column using `apply()`.

### Business Example

An e-commerce company may classify products based on price, sales amount, or customer spending.

`apply()` can be useful when the transformation requires custom logic that cannot be easily expressed using a simple vectorized operation.

### AI/ML Use Case

`apply()` can be used during data preprocessing and feature engineering when custom transformations or rules are required.

For simple numerical calculations, vectorized operations should generally be preferred because they are often more efficient.

````python
import pandas as pd

product_df = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Mouse"],
    "Price": [50000, 20000, 500]
})

def classify_price(price):
    if price > 30000:
        return "Expensive"
    else:
        return "Affordable"

product_df["Price_Category"] = product_df["Price"].apply(classify_price)

print(product_df)
````

## Step 3: Code Explanation

### Code Explanation

The DataFrame contains product names and prices.

The `classify_price()` function checks each price.

If the price is greater than 30000, it returns `"Expensive"`.

Otherwise, it returns `"Affordable"`.

The `apply()` method applies this function to every value in the Price column.

### What the Code Does

The code creates a new column called `Price_Category`.

The classification is:

- Laptop → 50000 → Expensive
- Phone → 20000 → Affordable
- Mouse → 500 → Affordable

### Why the Function/Method Is Used

- `pd.DataFrame()` → Creates the product dataset.
- `classify_price()` → Defines the custom classification rule.
- `.apply()` → Applies the custom function to each value in the Price column.
- `if` condition → Determines the price category.

### Expected Output

The output will be similar to:

       Product  Price Price_Category
0       Laptop  50000      Expensive
1        Phone  20000     Affordable
2        Mouse    500     Affordable

The new `Price_Category` column contains the result of the custom function.

### Where It Is Used in AI/ML or Data Analytics

`apply()` is useful for custom data transformations, feature engineering, and rule-based preprocessing.

For simple numerical operations, vectorization is generally preferred over both `apply()` and explicit Python loops.

# Performance Optimization - Memory Optimization

## Step 1: Understand the Concept

Memory Optimization is the process of reducing unnecessary memory usage while working with datasets.

Large datasets can consume a significant amount of RAM.

Pandas provides methods such as `memory_usage()` to check how much memory a DataFrame uses.

One common optimization technique is changing a column to a smaller suitable data type.

Why do we use Memory Optimization?

- To reduce RAM usage.
- To work efficiently with large datasets.
- To improve resource utilization.
- To make data processing more efficient.

`memory_usage(deep=True)` provides a more detailed memory usage calculation, especially for object and string data.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine storing items in a warehouse.

If we use a very large storage space for every small item, a lot of space will be wasted.

Instead, we can use storage space according to the actual size of each item.

Memory optimization works in a similar way by choosing suitable data types and reducing unnecessary memory usage.

### Simple Example

Suppose an Age column contains values between 18 and 60.

We do not need a very large integer data type to store these values.

A smaller suitable integer type can use less memory.

### Business Example

An e-commerce company may store millions of customer records.

If each column uses more memory than necessary, the entire dataset can consume a large amount of RAM.

Optimizing suitable columns can reduce memory usage.

### AI/ML Use Case

Large datasets are commonly used for machine learning.

Memory optimization can help data scientists load larger datasets into memory and perform preprocessing more efficiently.

````python
import pandas as pd

customer_df = pd.DataFrame({
    "Age": [22, 25, 30, 35, 40],
    "Name": ["Ravi", "Priya", "Arun", "Meena", "Karthik"]
})

print("Data Types Before Optimization:")
print(customer_df.dtypes)

print("\nMemory Usage Before Optimization:")
print(customer_df.memory_usage(deep=True))

# Convert Age to a smaller integer data type
customer_df["Age"] = customer_df["Age"].astype("int8")

print("\nData Types After Optimization:")
print(customer_df.dtypes)

print("\nMemory Usage After Optimization:")
print(customer_df.memory_usage(deep=True))
````

## Step 3: Code Explanation

### Code Explanation

The DataFrame contains customer names and ages.

First, the data types and memory usage are checked.

The Age column is then converted to the `int8` data type because the values are small and fit within the range of this data type.

The memory usage is checked again after the conversion.

### What the Code Does

The code:

1. Creates a customer DataFrame.
2. Checks the existing data types.
3. Checks memory usage.
4. Converts the Age column to `int8`.
5. Checks the data types again.
6. Checks memory usage after optimization.

### Why the Function/Method Is Used

- `memory_usage(deep=True)` → Checks detailed memory usage.
- `.astype("int8")` → Converts the column to a smaller integer data type.
- `dtypes` → Shows the data type of each column.

### Expected Output

The `Age` column will show `int8` after optimization.

The memory used by the Age column will generally be lower after converting it to a smaller suitable data type.

The exact memory values can vary depending on the Pandas version and system.

### Where It Is Used in AI/ML or Data Analytics

Memory optimization is useful when working with large datasets, especially during data preprocessing, feature engineering, and machine learning workflows where datasets may contain millions of rows.

# Performance Optimization - Category Data Type

## Step 1: Understand the Concept

The `category` data type is used for columns that contain a limited number of repeated categorical values.

Examples include:

- City
- Product Category
- Department
- Customer Type
- Status

Pandas stores the unique values as categories and represents the data using these categories.

Why do we use the Category data type?

- To reduce memory usage for suitable columns.
- To represent categorical data clearly.
- To improve efficiency for some categorical operations.
- To make the meaning of categorical columns explicit.

The Category data type is most useful when a column has a limited number of unique values compared with the total number of rows.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a company has 10,000 customer records, but the customers belong to only three cities:

- Chennai
- Madurai
- Coimbatore

Instead of treating every city name as completely separate text, we can maintain these three values as categories and refer to them efficiently.

### Simple Example

Suppose a City column contains:

| City |
|---|
| Chennai |
| Madurai |
| Chennai |
| Coimbatore |
| Madurai |

There are only three unique city values.

This type of column is a good candidate for the `category` data type.

### Business Example

An e-commerce company may have millions of orders but only a limited number of product categories such as:

- Electronics
- Clothing
- Grocery
- Furniture

Using the `category` data type can be useful for such repeated categorical values.

### AI/ML Use Case

Categorical columns are common in machine learning datasets.

Using the `category` data type can help manage categorical data efficiently during preprocessing and exploratory data analysis.

````python
import pandas as pd

customer_df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun", "Meena", "Karthik"],
    "City": ["Chennai", "Madurai", "Chennai", "Coimbatore", "Madurai"]
})

print("Data Type Before Optimization:")
print(customer_df["City"].dtype)

# Convert City to category data type
customer_df["City"] = customer_df["City"].astype("category")

print("\nData Type After Optimization:")
print(customer_df["City"].dtype)

print("\nCategories:")
print(customer_df["City"].cat.categories)
````

## Step 3: Code Explanation

### Code Explanation

The DataFrame contains customer names and cities.

Initially, the City column is stored as a regular text column.

The `astype("category")` method converts the City column to the Category data type.

The `.cat.categories` attribute displays the unique categories stored in the column.

### What the Code Does

The code:

1. Creates a customer DataFrame.
2. Checks the original data type of the City column.
3. Converts the City column to the Category data type.
4. Displays the available categories.

### Why the Function/Method Is Used

- `astype("category")` → Converts a column to the Category data type.
- `.cat.categories` → Displays the unique categories.
- `dtype` → Shows the data type of the column.

### Expected Output

The data type before conversion will typically be:

`object`

After conversion, it will be:

`category`

The categories will be:

- Chennai
- Coimbatore
- Madurai

### Where It Is Used in AI/ML or Data Analytics

The Category data type is useful for repeated categorical values in large datasets, such as cities, product categories, departments, and customer types. It can reduce memory usage and make categorical data easier to manage.

# Performance Optimization - Efficient Pandas Practices

## Step 1: Understand the Concept

Efficient Pandas practices are techniques used to make Pandas code cleaner, faster, and more memory-efficient.

Important practices include:

- Select only the columns that are required.
- Prefer vectorized operations over unnecessary loops.
- Use suitable data types.
- Use built-in Pandas methods for common operations.
- Avoid creating unnecessary copies of large DataFrames.
- Process large files in chunks when required.

Why do we use efficient Pandas practices?

- To reduce unnecessary processing.
- To reduce memory usage.
- To improve performance on large datasets.
- To write cleaner and more maintainable code.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine carrying items while moving to a new house.

If you carry everything, even things you do not need, it takes more time and space.

Instead, carrying only the required items makes the process faster and easier.

Efficient Pandas practices work in a similar way by processing only the data that is actually needed.

### Simple Example

Suppose a DataFrame contains:

| Name | Age | City | Salary |
|---|---:|---|---:|
| Ravi | 25 | Chennai | 30000 |
| Priya | 28 | Madurai | 35000 |
| Arun | 30 | Chennai | 40000 |

If we only need Name and Salary, we can select only those columns instead of processing the entire DataFrame.

### Business Example

An e-commerce company may have a dataset containing hundreds of columns.

For a sales report, only columns such as Product, Quantity, and Sales may be required.

Selecting only the required columns reduces unnecessary processing.

### AI/ML Use Case

Machine learning datasets can contain many features.

Efficient Pandas practices help with selecting required features, transforming data using vectorized operations, choosing suitable data types, and processing large datasets efficiently before model training.

````python
import pandas as pd

sales_df = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Tablet", "Mouse"],
    "Quantity": [2, 5, 3, 10],
    "Sales": [100000, 100000, 90000, 10000],
    "City": ["Chennai", "Madurai", "Chennai", "Coimbatore"]
})

# Select only the required columns
required_df = sales_df[["Product", "Sales"]]

# Use a vectorized operation
required_df["Discounted_Sales"] = required_df["Sales"] * 0.90

print(required_df)
````

## Step 3: Code Explanation

### Code Explanation

The DataFrame contains Product, Quantity, Sales, and City columns.

Only Product and Sales are required for this example, so we select those columns.

A vectorized multiplication is then used to calculate the discounted sales.

### What the Code Does

The code:

1. Creates a sales DataFrame.
2. Selects only the required Product and Sales columns.
3. Calculates discounted sales using a vectorized operation.
4. Stores the result in a new column.

The discount calculation is:

`Sales * 0.90`

This keeps 90% of the original sales value.

### Why the Function/Method Is Used

- `pd.DataFrame()` → Creates the sales dataset.
- `sales_df[["Product", "Sales"]]` → Selects only the required columns.
- `* 0.90` → Performs a vectorized calculation on the complete Sales column.
- `Discounted_Sales` → Stores the calculated values.

### Expected Output

The output will be similar to:

       Product   Sales  Discounted_Sales
0       Laptop  100000           90000.0
1        Phone  100000           90000.0
2       Tablet   90000           81000.0
3        Mouse   10000            9000.0

Only the required columns are processed, and the calculation is performed using a vectorized operation.

### Where It Is Used in AI/ML or Data Analytics

Efficient Pandas practices are used throughout data cleaning, preprocessing, feature engineering, exploratory data analysis, and machine learning workflows to process data efficiently and reduce unnecessary memory and computation.
